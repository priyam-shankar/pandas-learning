# Exploratory Data Analysis (EDA)

## 1. What is EDA?

**EDA stands for Exploratory Data Analysis.**

EDA is the process of exploring, analyzing, and understanding a dataset to discover patterns, identify errors, detect unusual values, and extract useful insights before applying Machine Learning.

In simple words, EDA means understanding our data before using it.

### Real-World Example

Imagine an e-commerce company like Amazon has a dataset containing product names, prices, ratings, and customer reviews.

The company wants to understand:
- Which products have low ratings?
- What is the average product rating?
- Are there any missing values?
- Are there duplicate records?
- Is there any relationship between product prices and ratings?

We investigate these questions using EDA.

*Note: Amazon is an illustrative example; the dataset below is sample data.*

---

## 2. Why Do We Need EDA?

EDA is important for the following reasons:

### 2.1 Understanding Data
EDA helps us understand the structure, features, size, and types of data available.

### 2.2 Finding Data Quality Issues
Real-world datasets may contain:
- Missing values
- Duplicate records
- Incorrect data types
- Invalid values

EDA helps us identify these problems.

### 2.3 Discovering Patterns
EDA helps us discover trends and patterns, such as products with low ratings or customers who purchase frequently.

### 2.4 Detecting Outliers
An outlier is a value that is unusually different from other observations.

For example, if most product prices are between ₹500 and ₹2,000, a price of ₹200,000 may need investigation.

An outlier is not always an error; it may represent a genuine expensive product.

### 2.5 Preparing Data for Machine Learning
EDA helps us make decisions about missing values, useful features, unusual observations, and appropriate preprocessing techniques before training an ML model.

---

## 3. EDA Workflow

A typical EDA workflow includes:

1. Collect the dataset.
2. Inspect the dataset structure.
3. Understand rows, columns, and data types.
4. Check for missing values.
5. Check for duplicate records.
6. Calculate descriptive statistics.
7. Identify outliers.
8. Explore distributions and relationships.
9. Visualize the data using charts.
10. Summarize findings and prepare the data for further analysis or Machine Learning.

EDA is an iterative process. We may return to earlier steps when we discover something unexpected.

---

## 4. Important EDA Terminology

| Term | Meaning |
|---|---|
| Dataset | A collection of data |
| Record / Row | One observation, such as one product |
| Feature / Column | A property, such as price or rating |
| Missing Value | A value that is unavailable |
| Duplicate | A repeated record |
| Outlier | An unusually different observation |
| Mean | The arithmetic average |
| Median | The middle value after sorting |
| Distribution | How values are spread across a dataset |
| Correlation | A measure of how two variables vary together |
| Data Cleaning | Identifying and correcting data quality issues |
| Data Visualization | Representing data using charts and graphs |

**Important:** Correlation does not imply causation. Two variables may be related without one causing the other.

## 7. Calculating the Average Rating

We can calculate the mean rating to understand the overall rating level of our sample products.

```python
ratings = []

for product in products:
    ratings.append(product["rating"])

average_rating = sum(ratings) / len(ratings)

print(average_rating)
```

**Output:**
```text
3.7
```

The average rating is 3.7.

### Formula

Mean = Sum of all values / Number of values

**Business Insight:** The average gives us a quick summary of product ratings. However, it does not reveal the complete distribution or explain why particular products receive low ratings.

## 9. Identifying Missing Values

Missing values occur when a field has no available value.

For example, let's make the Keyboard's rating missing.

```python
products[1]["rating"] = None
```

Now, we can identify missing ratings.

```python
for product in products:
    if product["rating"] is None:
        print("Missing rating:", product["name"])
```

**Output:**
```text
Missing rating: Keyboard
```

### How Should We Handle Missing Values?

Possible approaches include:
- Keep the missing value if it is meaningful.
- Remove affected records when justified.
- Fill missing values using an appropriate method.
- Investigate why the data is missing.

**Important:** Do not automatically replace every missing rating with zero. A zero rating and an unknown rating represent different things.

## 13. Common EDA Mistakes to Avoid

1. Starting Machine Learning before understanding the dataset.
2. Ignoring missing values.
3. Removing outliers without investigation.
4. Treating every unusual value as an error.
5. Assuming correlation proves causation.
6. Looking only at averages while ignoring distributions.
7. Making conclusions from a very small sample without acknowledging its limitations.
8. Drawing business conclusions without sufficient supporting evidence.

# Pandas Series — Concepts and Industry Applications

## 1. What Is a Pandas Series?

A **Pandas Series** is a one-dimensional labeled array that stores data along with an associated index.

A Series has three important components:

- **Values:** The actual information stored in the Series.
- **Index:** Labels used to identify and access individual values.
- **Data type (`dtype`):** Describes the type of data represented by the Series.

Pandas Series is useful when working with a single sequence of labeled data.

## 2. Default Index and Custom Index

By default, Pandas assigns integer index labels starting from zero.

A custom index allows meaningful labels such as employee names, product IDs, dates, or city names.

For example, in an employee age Series:

- Index label: `Priyam`
- Corresponding value: `21`

Here, `Priyam` identifies the data, while `21` is the actual value.

**Industry application:** Custom indexes can make employee records, product metrics, and time-based observations easier to access.

## 3. Accessing Series Elements

Series elements can be accessed using positions or labels.

### Position-Based Access

Position-based access identifies an element by its numerical position. Positions start at zero.

### Label-Based Access

Label-based access identifies an element using its index label.

The main distinction is:

- `.iloc` accesses data by position.
- `.loc` accesses data by label.

**Industry application:** A data analyst might retrieve a particular employee's salary, a product's price, or a specific day's sales using the appropriate index.

## 4. Characteristics of a Series

### 4.1 Data Types and Homogeneity

A Series generally represents its values using a common data type.

However, Pandas can also store mixed Python objects in a Series. In such cases, the resulting data type may be `object`.

The resulting dtype depends on the values and Pandas' type-inference rules.

**Why it matters:** Consistent data types make numerical analysis and Machine Learning preprocessing easier.

### 4.2 Vectorized Operations

Vectorization allows operations to be applied to Series values without explicitly writing a Python loop for each element.

For example, vectorized addition can be used to calculate updated sales, revised salaries, or adjusted prices.

Pandas aligns Series by index labels during arithmetic operations. If the indexes differ, the resulting values may not match by position, and missing values can appear.

**Industry application:** Calculating daily revenue, applying price adjustments, comparing metrics, and processing numerical features.

### 4.3 Missing Values

Missing values represent information that is unavailable or not recorded.

Pandas commonly uses `NaN` for missing numerical data, although other missing-value representations also exist.

A missing value is not necessarily zero.

For example, an absent customer purchase amount does not automatically mean that the customer spent nothing.

**Industry application:** Missing-value analysis is important when preparing real-world data for reporting and Machine Learning.

### 4.4 Mutability

A Series is mutable, meaning existing values can be changed.

However, some Series operations return a new Series rather than modifying the original object.

### 4.5 Removing Elements

The `drop()` method returns a Series with specified index labels removed. By default, it does not modify the original Series.

The resulting Series normally retains the remaining original index labels.

**Important:** Dropping index label `0` means removing the label `0`, not necessarily removing an element based on its current position.

## 5. Series Versus DataFrame

A Series is one-dimensional and generally represents a single labeled sequence.

A DataFrame is two-dimensional and organizes data into rows and columns.

For example:

- A Series can represent employee salaries.
- A DataFrame can represent employee names, salaries, departments, and joining dates together.

Both are important Pandas data structures used in data analysis.

## 6. Real-World Industry Applications

### E-commerce

A Series can represent daily sales, product prices, product ratings, or quantities sold.

### Human Resources

A Series can represent employee salaries, attendance counts, or individual performance metrics.

### Finance

A Series can represent daily stock prices, expenses, or revenue over time.

### Healthcare

A Series can represent a single numerical measurement across observations, subject to appropriate privacy and data-handling requirements.

### Machine Learning

A Series can represent a single feature or a target variable. It is also useful when performing numerical operations and inspecting data quality.

A Series alone is not a complete Machine Learning workflow; real projects often require DataFrames, preprocessing, validation, and additional analysis.

## 7. Important Takeaways

- A Series is a one-dimensional labeled array.
- Values and index labels are different components.
- Pandas supports default and custom indexes.
- `.iloc` is position-based, while `.loc` is label-based.
- Vectorized operations simplify numerical analysis.
- Pandas aligns Series using index labels during arithmetic.
- Missing values require careful interpretation.
- Existing values can be modified.
- Operations such as `drop()` return a new Series by default.
- Series is suitable for a single labeled sequence, while DataFrame organizes multiple columns.

- Smartwatch: ₹8,000

### Hints

- Explore the Pandas Series method `sort_values()`.

# Practice 9: Frequency Analysis with value_counts()

## Scenario
You work for a food-delivery company. You need to find which food category customers order most frequently.

### Given Data
Orders: `"Pizza", "Burger", "Pizza", "Biryani", "Burger", "Pizza", "Dosa", "Burger", "Biryani", "Pizza"`

### Tasks
1. Create a Series named `food_orders`.
2. Display all orders.
3. Count the occurrences of each food category.
4. Identify the most frequently ordered category.
5. Display the frequency counts in descending order.

### Hints
- Explore the `value_counts()` method.
- Think about how to retrieve the category with the highest frequency from the resulting Series.

**Industry Insight:** Food-delivery platforms use order-frequency analysis to understand demand, plan inventory, and identify popular dishes.


````python
food_orders = pd.Series([
    "Pizza", "Burger", "Pizza", "Biryani", "Burger",
    "Pizza", "Dosa", "Burger", "Biryani", "Pizza"
])

# Display all orders
print("All Food Orders:")
print(food_orders)

# Count each food category
print("\nFood Order Frequencies:")
print(food_orders.value_counts())

# Identify the most frequently ordered food
print("\nMost Frequently Ordered Food:")
print(food_orders.value_counts().idxmax())

# Display frequencies in descending order
print("\nFrequencies in Descending Order:")
print(food_orders.value_counts(ascending=False))
````
