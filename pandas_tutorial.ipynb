# Exploratory Data Analysis (EDA)

## 1. What is EDA?

**EDA stands for Exploratory Data Analysis.**

EDA is the process of exploring, analyzing, and understanding a dataset to discover patterns, identify errors, detect unusual values, and extract useful insights before applying Machine Learning.

In simple words, EDA means understanding our data before using it.

### Real-World Example

Imagine an e-commerce company like Amazon has a dataset containing product names, prices, ratings, and customer reviews.

The company wants to understand:
- Which products have low ratings?
- What is the average product rating?
- Are there any missing values?
- Are there duplicate records?
- Is there any relationship between product prices and ratings?

We investigate these questions using EDA.

*Note: Amazon is an illustrative example; the dataset below is sample data.*

---

## 2. Why Do We Need EDA?

EDA is important for the following reasons:

### 2.1 Understanding Data
EDA helps us understand the structure, features, size, and types of data available.

### 2.2 Finding Data Quality Issues
Real-world datasets may contain:
- Missing values
- Duplicate records
- Incorrect data types
- Invalid values

EDA helps us identify these problems.

### 2.3 Discovering Patterns
EDA helps us discover trends and patterns, such as products with low ratings or customers who purchase frequently.

### 2.4 Detecting Outliers
An outlier is a value that is unusually different from other observations.

For example, if most product prices are between ₹500 and ₹2,000, a price of ₹200,000 may need investigation.

An outlier is not always an error; it may represent a genuine expensive product.

### 2.5 Preparing Data for Machine Learning
EDA helps us make decisions about missing values, useful features, unusual observations, and appropriate preprocessing techniques before training an ML model.

---

## 3. EDA Workflow

A typical EDA workflow includes:

1. Collect the dataset.
2. Inspect the dataset structure.
3. Understand rows, columns, and data types.
4. Check for missing values.
5. Check for duplicate records.
6. Calculate descriptive statistics.
7. Identify outliers.
8. Explore distributions and relationships.
9. Visualize the data using charts.
10. Summarize findings and prepare the data for further analysis or Machine Learning.

EDA is an iterative process. We may return to earlier steps when we discover something unexpected.

---

## 4. Important EDA Terminology

| Term | Meaning |
|---|---|
| Dataset | A collection of data |
| Record / Row | One observation, such as one product |
| Feature / Column | A property, such as price or rating |
| Missing Value | A value that is unavailable |
| Duplicate | A repeated record |
| Outlier | An unusually different observation |
| Mean | The arithmetic average |
| Median | The middle value after sorting |
| Distribution | How values are spread across a dataset |
| Correlation | A measure of how two variables vary together |
| Data Cleaning | Identifying and correcting data quality issues |
| Data Visualization | Representing data using charts and graphs |

**Important:** Correlation does not imply causation. Two variables may be related without one causing the other.

## 7. Calculating the Average Rating

We can calculate the mean rating to understand the overall rating level of our sample products.

```python
ratings = []

for product in products:
    ratings.append(product["rating"])

print(average_rating)
```

```text

```

The average rating is 3.7.

### Formula

Mean = Sum of all values / Number of values

**Business Insight:** The average gives us a quick summary of product ratings. However, it does not reveal the complete distribution or explain why particular products receive low ratings.

## 9. Identifying Missing Values

Missing values occur when a field has no available value.

For example, let's make the Keyboard's rating missing.

```python

```

Now, we can identify missing ratings.

```python
for product in products:

        print("Missing rating:", product["name"])
```

**Output:**
```text

```

### How Should We Handle Missing Values?

Possible approaches include:
- Keep the missing value if it is meaningful.
- Remove affected records when justified.
- Fill missing values using an appropriate method.
- Investigate why the data is missing.

**Important:** Do not automatically replace every missing rating with zero. A zero rating and an unknown rating represent different things.

## 13. Common EDA Mistakes to Avoid

1. Starting Machine Learning before understanding the dataset.
2. Ignoring missing values.
3. Removing outliers without investigation.
4. Treating every unusual value as an error.
5. Assuming correlation proves causation.
6. Looking only at averages while ignoring distributions.
7. Making conclusions from a very small sample without acknowledging its limitations.
8. Drawing business conclusions without sufficient supporting evidence.

# Pandas Series — Concepts and Industry Applications

## 1. What Is a Pandas Series?

A **Pandas Series** is a one-dimensional labeled array that stores data along with an associated index.

A Series has three important components:

- **Values:** The actual information stored in the Series.
- **Index:** Labels used to identify and access individual values.
- **Data type (`dtype`):** Describes the type of data represented by the Series.

Pandas Series is useful when working with a single sequence of labeled data.

## 2. Default Index and Custom Index

By default, Pandas assigns integer index labels starting from zero.

A custom index allows meaningful labels such as employee names, product IDs, dates, or city names.

For example, in an employee age Series:

- Index label: `Priyam`
- Corresponding value: `21`

Here, `Priyam` identifies the data, while `21` is the actual value.

**Industry application:** Custom indexes can make employee records, product metrics, and time-based observations easier to access.

## 3. Accessing Series Elements

Series elements can be accessed using positions or labels.

### Position-Based Access

Position-based access identifies an element by its numerical position. Positions start at zero.

### Label-Based Access

Label-based access identifies an element using its index label.

The main distinction is:

- `.iloc` accesses data by position.
- `.loc` accesses data by label.

**Industry application:** A data analyst might retrieve a particular employee's salary, a product's price, or a specific day's sales using the appropriate index.

## 4. Characteristics of a Series

### 4.1 Data Types and Homogeneity

A Series generally represents its values using a common data type.

However, Pandas can also store mixed Python objects in a Series. In such cases, the resulting data type may be `object`.

The resulting dtype depends on the values and Pandas' type-inference rules.

**Why it matters:** Consistent data types make numerical analysis and Machine Learning preprocessing easier.

### 4.2 Vectorized Operations

Vectorization allows operations to be applied to Series values without explicitly writing a Python loop for each element.

For example, vectorized addition can be used to calculate updated sales, revised salaries, or adjusted prices.

Pandas aligns Series by index labels during arithmetic operations. If the indexes differ, the resulting values may not match by position, and missing values can appear.

**Industry application:** Calculating daily revenue, applying price adjustments, comparing metrics, and processing numerical features.

### 4.3 Missing Values

Missing values represent information that is unavailable or not recorded.

Pandas commonly uses `NaN` for missing numerical data, although other missing-value representations also exist.

A missing value is not necessarily zero.

For example, an absent customer purchase amount does not automatically mean that the customer spent nothing.

**Industry application:** Missing-value analysis is important when preparing real-world data for reporting and Machine Learning.

### 4.4 Mutability

A Series is mutable, meaning existing values can be changed.

However, some Series operations return a new Series rather than modifying the original object.

### 4.5 Removing Elements

The `drop()` method returns a Series with specified index labels removed. By default, it does not modify the original Series.

The resulting Series normally retains the remaining original index labels.

**Important:** Dropping index label `0` means removing the label `0`, not necessarily removing an element based on its current position.

## 5. Series Versus DataFrame

A Series is one-dimensional and generally represents a single labeled sequence.

A DataFrame is two-dimensional and organizes data into rows and columns.

For example:

- A Series can represent employee salaries.
- A DataFrame can represent employee names, salaries, departments, and joining dates together.

Both are important Pandas data structures used in data analysis.

## 6. Real-World Industry Applications

### E-commerce

A Series can represent daily sales, product prices, product ratings, or quantities sold.

### Human Resources

A Series can represent employee salaries, attendance counts, or individual performance metrics.

### Finance

A Series can represent daily stock prices, expenses, or revenue over time.

### Healthcare

A Series can represent a single numerical measurement across observations, subject to appropriate privacy and data-handling requirements.

### Machine Learning

A Series can represent a single feature or a target variable. It is also useful when performing numerical operations and inspecting data quality.

A Series alone is not a complete Machine Learning workflow; real projects often require DataFrames, preprocessing, validation, and additional analysis.

## 7. Important Takeaways

- A Series is a one-dimensional labeled array.
- Values and index labels are different components.
- Pandas supports default and custom indexes.
- `.iloc` is position-based, while `.loc` is label-based.
- Vectorized operations simplify numerical analysis.
- Pandas aligns Series using index labels during arithmetic.
- Missing values require careful interpretation.
- Existing values can be modified.
- Operations such as `drop()` return a new Series by default.
- Series is suitable for a single labeled sequence, while DataFrame organizes multiple columns.

# Pandas DataFrame — Introduction and Creation

## 1. What is a DataFrame?

A **DataFrame** is a two-dimensional, labeled tabular data structure in Pandas. It stores data in rows and columns, similar to an Excel spreadsheet or a database table.

- **Rows** represent individual records.
- **Columns** represent attributes or features.
- **Row labels (index)** identify rows.
- **Column labels** identify columns.
- Each column of a DataFrame is a Pandas **Series**.

### Example

Consider a student dataset with three columns: `Name`, `Age`, and `GPA`.

| Index | Name | Age | GPA |
|---:|---|---:|---:|
| 0 | Priyam | 21 | 7.7 |
| 1 | Rupak | 23 | 7.5 |
| 2 | Amit | 23 | 5.2 |

Here, each row represents one student, while each column represents a particular attribute.

## 2. Creating a DataFrame

Pandas provides several ways to create a DataFrame.

### Method 1: Dictionary of Lists

A dictionary of lists is a common way to create a DataFrame.

- Dictionary keys become column names.
- Dictionary values provide the data for each column.
- Each list generally needs to have the same length.

**Use case:** Creating structured datasets such as student records, employee information, or sales reports.

### Method 2: List of Lists

A DataFrame can also be created from a list of lists.

- Each inner list represents one row.
- Values are arranged according to their position.
- The `columns` parameter assigns meaningful column names.

**Use case:** Converting manually collected records or rows of data into a tabular structure.

### Method 3: NumPy Array

A NumPy array can be converted into a Pandas DataFrame.

- A two-dimensional array maps naturally to rows and columns.
- The `columns` parameter assigns column labels.
- If column names are not supplied, Pandas generates default integer column labels.

**Use case:** Converting numerical data from scientific computing, simulations, or machine learning preprocessing into a labeled table.

## 3. Understanding DataFrame Labels

A DataFrame has two types of labels.

### Row Labels — `df.index`

The `index` attribute returns the row labels.

When an index is not explicitly provided, Pandas normally assigns a default integer index starting from `0`.

For example, a DataFrame with three rows typically has an index equivalent to `RangeIndex(start=0, stop=3, step=1)`.

### Column Labels — `df.columns`

The `columns` attribute returns the column names.

These labels help identify and access specific attributes in the dataset.

For example, a student DataFrame might have column labels such as `Name`, `Age`, and `GPA`.

**Important:** `df.index` and `df.columns` return label information; they do not return the actual cell values.

## 4. Series vs DataFrame

|---|---|---|
| Dimensions | One-dimensional | Two-dimensional |
| Structure | One labeled sequence | Rows and columns |
| Labels | Index | Row index and column labels |
| Typical use | One variable | Multiple related variables |

A DataFrame column is a Series. For example, selecting the `Age` column from a student DataFrame normally returns a Series.

# Reading External Files in Pandas

## 1. Why Do We Import Data from External Files?

In real-world projects, data is rarely written manually inside Python code. Organisations usually store data in external files, databases, cloud storage, or APIs.

For example:
- An HR department stores employee information in a CSV file.
- An e-commerce company stores product details in a JSON file.
- A bank stores transaction records in a database.
- A data scientist imports historical data to train a machine learning model.

Pandas allows us to load external data into a DataFrame so that we can inspect, clean, transform, analyse, and prepare it for further use.

**Basic workflow:**

External file → Pandas → DataFrame → Data cleaning and analysis → Insights or machine learning

## 2. What Is a CSV File?

**CSV stands for Comma-Separated Values.**

It is a plain-text file format used to store tabular data. Each line generally represents a row, and commas separate the values within that row.

Example of `employee_data.csv`:

```csv
Name,Age,Department,Salary
Priyam,21,Engineering,45000
Rupak,23,Marketing,38000
Amit,24,Finance,50000
```

In this example:
- The first row contains column headers.
- Each subsequent row represents an employee.
- Each comma separates one field from another.

CSV files are widely used because they are simple, lightweight, and supported by spreadsheets, databases, and data analysis tools.

### Reading a CSV File

Pandas provides the `pd.read_csv()` function to read CSV files.

Conceptually, this function:
1. Opens and reads the CSV file.
2. Interprets the rows and columns.
3. Uses the first row as column names by default.
4. Converts the data into a Pandas DataFrame.

The resulting DataFrame can then be used for operations such as filtering employees, calculating average salaries, or analysing departments.

### Important Parameters of `pd.read_csv()`

| Parameter | Purpose |
|---|---|
| `filepath_or_buffer` | Specifies the file path or readable source. |
| `sep` | Specifies the delimiter; the default is a comma. |
| `header` | Specifies which row contains column names. |
| `names` | Provides custom column names. |
| `index_col` | Specifies a column to use as the row index. |
| `usecols` | Loads only selected columns. |
| `nrows` | Reads only a specified number of rows. |
| `skiprows` | Skips specified rows when reading. |
| `encoding` | Specifies the file's text encoding. |
| `na_values` | Identifies additional values that should be treated as missing. |

You do not need to memorise every parameter immediately. Start with the basic function and learn the others when a project requires them.

## 3. What Is a JSON File?

**JSON stands for JavaScript Object Notation.**

It is a text-based format used to represent structured data. JSON is particularly common in web applications, REST APIs, configuration files, and communication between services.

Unlike CSV, JSON can represent nested objects and arrays, making it suitable for hierarchical data.

Example of `employee_data.json`:

```json
[
  {
    "Name": "Priyam",
    "Age": 21,
    "Department": "Engineering",
    "Salary": 45000
  },
  {
    "Name": "Rupak",
    "Age": 23,
    "Department": "Marketing",
    "Salary": 38000
  }
]
```

Here:
- Square brackets `[]` represent an array.
- Curly braces `{}` represent objects.
- Keys such as `"Name"` identify fields.
- Values store the corresponding information.
- Each object represents an employee.

JSON supports strings, numbers, booleans, null values, arrays, and nested objects.

### Reading a JSON File

Pandas provides `pd.read_json()` to load supported JSON data into a DataFrame.

Conceptually, this function:
1. Reads the JSON file.
2. Parses its structure.
3. Converts compatible records or data structures into a Pandas object, commonly a DataFrame.
4. Makes the data available for analysis.

The exact result depends on the JSON structure. A JSON file containing a list of employee objects is a common example that becomes a DataFrame.

### Important Parameters of `pd.read_json()`

| Parameter | Purpose |
|---|---|
| `path_or_buf` | Specifies the file path or readable source. |
| `orient` | Describes how the JSON data is structured. |
| `typ` | Specifies whether to return a Series or DataFrame when applicable. |
| `lines` | Indicates whether each line contains a separate JSON object. |
| `encoding` | Specifies the text encoding. |

For example, some APIs return **JSON Lines (NDJSON)**, where each line is a separate JSON object. This format may require `lines=True`.

## 4. CSV vs JSON

| Feature | CSV | JSON |
|---|---|---|
| Full form | Comma-Separated Values | JavaScript Object Notation |
| Structure | Rows and columns | Objects, arrays and nested structures |
| Common use | Tabular datasets | APIs and structured application data |
| Nested data | Not naturally supported | Supported |
| Pandas reader | `pd.read_csv()` | `pd.read_json()` |

**Practical rule:** CSV is often convenient for spreadsheet-like datasets, while JSON is common when data comes from web APIs or contains nested structures. The right choice depends on the source and structure of the data.

## 5. Understanding File Paths

When you pass a filename to a Pandas reading function, Python needs to locate that file.

### Relative Path

A relative path is interpreted from the current working directory.

For example, `employee_data.csv` means Python will look for that file in the current working directory.

### Absolute Path

Both `pd.read_csv()` and `pd.read_json()` commonly return a Pandas DataFrame when the input contains tabular records.

## 7. Common Errors and Their Causes

### `FileNotFoundError`

**Cause:** Python cannot find the file at the specified path.

**What to check:**
- Is the filename correct?
- Is the extension correct?
- Is the file in the current working directory?
- Does the relative or absolute path point to the right location?

### `ParserError`

**Cause:** A CSV file may contain malformed rows or an unexpected structure.

**What to check:**
- Are the delimiters consistent?
- Are rows formatted correctly?
- Are quoted fields properly closed?

### `UnicodeDecodeError`

**Cause:** The file's text encoding does not match the encoding used to read it.

**What to check:** Determine the file's encoding and use the appropriate `encoding` parameter when necessary.

### `ValueError` or JSON parsing errors

**Cause:** The file may contain invalid JSON or a structure that the selected reading method cannot interpret as expected.

**What to check:**
- Are quotes, commas and brackets valid?
- Does the file contain standard JSON or JSON Lines?
- Does the JSON structure match the intended reading options?

# Pandas DataFrame Methods — EDA Basics

## 1. What Is EDA?

**EDA (Exploratory Data Analysis)** is the process of understanding a dataset before performing further analysis or building a machine learning model.

It helps us understand:
- How much data is available.
- What columns are present.
- Which data types are used.
- Whether values are missing.
- How numerical values are distributed.
- How many distinct values exist in each column.

Pandas provides several methods and attributes to perform these initial checks efficiently.

## 2. `head()` — View the First Rows

The `head()` method displays the first few rows of a DataFrame.

**Syntax:** `df.head(n)`

- `n` specifies the number of rows to display.
- By default, `n = 5`.
- It is useful for checking the structure and values at the beginning of a dataset.

**Example:** `df.head(3)` displays the first three rows.

**Real-world use:** Quickly inspect the first few employee records after loading a CSV file.

## 3. `tail()` — View the Last Rows

The `tail()` method displays the last few rows of a DataFrame.

**Syntax:** `df.tail(n)`

- By default, it displays the last five rows.
- Passing a number displays that many rows from the end.
- It helps inspect the most recent records or check the end of a loaded dataset.

**Example:** `df.tail(2)` displays the last two rows.

**Real-world use:** Check the latest transactions in a dataset sorted by transaction date.

**Important:** `head()` and `tail()` display existing rows; they do not sort or modify the DataFrame.

## 4. `sample()` — Select Random Rows

The `sample()` method returns a random sample of rows or columns from a DataFrame.

**Syntax:** `df.sample(n)`

- By default, `n = 1`, so one row is sampled.
- `df.sample(5)` selects five random rows, not the row at index 5.
- The selected rows can change between calls because the sampling is random.
- If reproducible results are needed, use the `random_state` parameter.
- By default, sampling is performed without replacement, so the requested sample size cannot exceed the number of available rows.

**Example:** `df.sample(5, random_state=42)` selects five rows reproducibly when the dataset remains unchanged.

**Real-world use:** Inspect a few random customer records to understand whether the data looks consistent across the dataset.

## 5. `info()` — Get a Structural Summary

The `info()` method displays a concise summary of a DataFrame.

**Syntax:** `df.info()`

It provides information such as:
- Number of rows and columns.
- Index range or index details.
- Column names.
- Number of non-null values in each column.
- Data type of each column.
- Approximate memory usage.

**Real-world use:** After importing a CSV file, check whether salary columns are numerical and whether any columns contain missing values.

**Important:** `info()` prints a summary and normally returns `None`. It is not a random summary, and it does not display all the actual data values.

## 6. `shape` — Find the Dataset Dimensions

`shape` is an **attribute**, not a method. Therefore, it does not use parentheses.

**Syntax:** `df.shape`

It returns a tuple in the form:

`(number_of_rows, number_of_columns)`

For example, `(1000, 8)` means the DataFrame has 1,000 rows and 8 columns.

**Real-world use:** Check the size of a dataset before processing it or splitting it into training and testing sets.

**Important:** The tuple order is rows first, columns second.

## 7. `describe()` — Generate Statistical Summaries

The `describe()` method calculates descriptive statistics for selected columns.

**Syntax:** `df.describe()`

For numerical columns, the output typically includes:

| Statistic | Meaning |
|---|---|
| `count` | Number of non-missing values |
| `mean` | Arithmetic average |
| `std` | Standard deviation, measuring spread |
| `min` | Minimum value |
| `25%` | First quartile |
| `50%` | Median, or second quartile |
| `75%` | Third quartile |
| `max` | Maximum value |

By default, `df.describe()` summarises numerical columns. You can use `df.describe(include="all")` to request summaries for all column types, although the statistics shown depend on the data type.

**Real-world use:** Understand employee salary ranges, average salaries, and whether unusually high or low values may need further investigation.

**Important:** `count` is the count of non-missing values, not necessarily the total number of rows. Standard deviation measures variation; it is not the same as the range.

## 8. `columns` — View Column Labels

`columns` is an attribute that returns the column labels of a DataFrame.

**Syntax:** `df.columns`

It returns a Pandas Index object containing the column names.

For example, a DataFrame might have the labels `Name`, `Age`, `Department`, and `Salary`.

**Real-world use:** Verify column names before selecting specific fields or preparing features for a machine learning model.

**Important:** Use `df.columns`, not `df.columns()`, because `columns` is an attribute rather than a method.

## 9. `nunique()` — Count Distinct Values

The `nunique()` method counts the number of unique values in each column by default.

**Syntax:** `df.nunique()`

- For a DataFrame, it returns the distinct-value count for each column.
- By default, `dropna=True`, so missing values (`NaN` or equivalent missing entries) are not counted as unique values.
- `df.nunique(axis=0)` counts distinct values column-wise.
- `df.nunique(axis=1)` counts distinct values row-wise.

**Real-world use:** Find out how many unique departments exist in an employee dataset or how many different product categories appear in sales data.

**Important:** `nunique()` returns a count, not the list of distinct values. To retrieve the distinct values from one Series, use `unique()` on that Series.

## 10. Method vs Attribute

A common beginner mistake is confusing methods with attributes.

| Type | Examples | Usage |
|---|---|---|
| Methods | `head()`, `tail()`, `sample()`, `info()`, `describe()`, `nunique()` | Usually called with parentheses |
| Attributes | `shape`, `columns`, `index` | Accessed without parentheses |

Methods perform an operation or return a result. Attributes provide information about an object's properties.

## 11. Recommended EDA Workflow

When a new dataset arrives, a practical initial inspection sequence is:

1. Use `head()` to inspect the first records.
2. Use `shape` to understand the dataset's dimensions.
3. Use `columns` to inspect column labels.
4. Use `info()` to examine data types and non-null counts.
5. Use `describe()` to understand numerical distributions.
6. Use `sample()` to inspect randomly selected records.
7. Use `tail()` to inspect the last records.
8. Use `nunique()` to check the variety of values in each column.

This is a flexible workflow, not a mandatory order. The appropriate checks depend on the dataset and the questions being investigated.

- `info()` summarises structure, data types, non-null counts, and memory usage.

# Pandas: Indexing and Selecting Data

## 1. What Is Indexing and Selecting?

**Indexing** means accessing specific data using its label or position. **Selecting** means choosing particular rows, columns, or individual values from a DataFrame.

In real-world projects, a dataset may contain thousands of rows and dozens of columns. We usually select only the information needed for analysis.

For example, an air-quality dataset may contain country, city, latitude, longitude, and AQI (Air Quality Index). We might need only city names, the first few records, or the AQI value of a particular city.

Pandas provides several ways to access this data:

- Selecting columns using column labels.
- Selecting rows using `.loc` and `.iloc`.
- Selecting individual cells using `.at` and `.iat`.
- Using `.copy()` to create an independent copy of selected data.

## 2. Selecting Columns

### A. Selecting a Single Column

Use the column name inside square brackets.

**Syntax:** `df["column_name"]`

Example:

```python
df_aqi["country"]
```

This selects the `country` column and normally returns a Pandas **Series**.

Column names are case-sensitive. For example, `"country"` and `"Country"` are different labels.

### B. Selecting Multiple Columns

To select multiple columns, pass a list of column names inside square brackets.

**Syntax:** `df[["column1", "column2"]]`

Example:

```python
df_aqi[["city", "aqi"]]
```

This returns a DataFrame containing only the selected columns, in the order specified.

**Remember the difference:**

- `df["city"]` → Series.
- `df[["city"]]` → DataFrame.
- `df[["city", "aqi"]]` → DataFrame with two columns.

## 3. Selecting Rows Using `.loc`

`.loc` is primarily a **label-based indexer**. It selects rows and columns using their labels.

### A. Selecting One Row

**Syntax:** `df.loc[row_label]`

Example:

```python
df_aqi.loc[0]
```

This selects the row whose index label is `0`.

The label does not necessarily mean the first physical row. If the DataFrame has a custom index, `.loc[0]` searches for the label `0`.

### B. Selecting Multiple Rows

You can select a range of row labels using slicing.

Example:

```python
df_aqi.loc[0:2]
```

With the usual integer index, this selects the rows labelled `0`, `1`, and `2`.

**Important:** `.loc` includes the ending label in a slice when that label is present.

This differs from ordinary Python slicing and `.iloc` slicing.

## 4. Selecting Rows Using `.iloc`

`.iloc` is a **position-based indexer**. It selects data using zero-based integer positions, regardless of the actual row labels.

### A. Selecting One Row

**Syntax:** `df.iloc[position]`

Example:

```python
df_aqi.iloc[2]
```

This selects the third row because positions start at zero.

### B. Selecting Multiple Rows

Example:

```python
df_aqi.iloc[0:3]
```

This selects positions `0`, `1`, and `2`.

**Important:** The ending position in `.iloc` slicing is exclusive.

| Expression | Meaning |
|---|---|
| `df.iloc[0]` | First row |
| `df.iloc[2]` | Third row |
| `df.iloc[0:3]` | First three rows |
| `df.iloc[0:2]` | First two rows |

### `.loc` vs `.iloc`

| Feature | `.loc` | `.iloc` |
|---|---|---|
| Access based on | Labels | Integer positions |
| Row selector | Index labels | Row positions |
| Column selector | Column labels | Column positions |
| Slice ending | Inclusive | Exclusive |

**Easy way to remember:** `.loc` uses labels; `.iloc` uses integer positions.

## 5. Selecting Rows and Columns Together

Sometimes we need specific columns from specific rows. Both `.loc` and `.iloc` support this.

The general pattern is:

`df.loc[row_selector, column_selector]`

or

`df.iloc[row_selector, column_selector]`

The comma separates the row selection from the column selection.

### A. Using `.loc`

Select the AQI value from the row labelled `0`:

```python
df_aqi.loc[0, "aqi"]
```

Select the city and AQI columns from that row:


Select those columns for rows labelled `0` through `2`:

```python
df_aqi.loc[0:2, ["city", "aqi"]]
```

The row-label slice includes the ending label `2`.

### B. Using `.iloc`

With `.iloc`, both row and column selections use integer positions.

Suppose the column positions are:

| Position | Column |
|---:|---|
| 0 | First column |
| 1 | Country |
| 2 | City |
| 3 | Latitude |
| 4 | Longitude |
| 11 | AQI |

The actual positions must be checked against the DataFrame's column order.

Select the third column from the first two rows:

```python
df_aqi.iloc[0:2, 2]
```

Select the columns at positions `1`, `2`, and `11` for the first three rows:

```python
df_aqi.iloc[0:3, [1, 2, 11]]
```

You can also select consecutive column positions using a slice. For example, `[1, 2, 3, 4]` can be replaced with `1:5`.

### Why Does `.iloc` Reject Column Names?

This expression is invalid:

```python
df_aqi.iloc[0:2, ["city", "aqi"]]
```

It raises an indexing error because `.iloc` expects integer positions, not column labels.

Use `.loc` when selecting by column names, or find the column positions and use `.iloc`.

To inspect the column labels and their order:

```python

```

## 6. Accessing a Single Cell Using `.at` and `.iat`

When you need only one scalar value from a DataFrame, `.at` and `.iat` provide direct access.

### A. `.at` — Label-Based Scalar Access

**Syntax:** `df.at[row_label, column_label]`

Example:

```python
df_aqi.at[0, "city"]
```

This accesses the value in the row labelled `0` and the column labelled `"city"`.

### B. `.iat` — Position-Based Scalar Access

**Syntax:** `df.iat[row_position, column_position]`

Example:

```python
df_aqi.iat[0, 2]
```

This accesses the value at the first row and third column.

### `.at` vs `.iat`

| Feature | `.at` | `.iat` |
|---|---|---|
| Selection | Labels | Integer positions |
| Row argument | Row label | Row position |
| Column argument | Column label | Column position |
| Best suited for | One scalar value | One scalar value |

Both methods are designed for accessing a single value. For selecting multiple rows or columns, `.loc` and `.iloc` are generally more appropriate.

## 7. Views, Copies, and `.copy()`

This is an important concept when modifying selected data.

### What Is a View?

A view is an object that refers to or shares underlying data with another object. Depending on the operation and Pandas version, selecting data may involve shared data or a separate object.

### What Is a Copy?

A copy contains independent data for the selected values. Changes to that copied data do not modify the original data through the copy.

### Why Use `.copy()`?

Suppose we want to extract the city column and change one of its values without changing the original DataFrame.

```python
cities = df_aqi["city"].copy()
```

The `.copy()` method creates an independent copy of the selected Series.

You can now modify the copied Series:

```python
cities.iloc[0] = "Mumbai"
```

This changes the value in `cities`, not the corresponding value in `df_aqi`.

### Important Modern Pandas Behaviour

Do not assume that selecting a column always returns a view or always returns a copy. The behaviour depends on the operation and Pandas version; Copy-on-Write also affects how changes are handled in modern Pandas.

**Best practice:** If you intend to modify selected data independently, explicitly use `.copy()`. If you intend to modify the original DataFrame, perform the assignment directly on the DataFrame using `.loc` or `.iloc`.

For example, to update the city in the original DataFrame:

```python
df_aqi.loc[0, "city"] = "Mumbai"
```

This explicitly assigns the new value to the original DataFrame.

## 8. Common Mistakes to Avoid

1. **Incorrect column spelling or capitalization:** Column labels must match the actual names.
2. **Confusing `.loc` and `.iloc`:** `.loc` uses labels; `.iloc` uses positions.
3. **Forgetting slice boundaries:** `.loc` includes the ending label, while `.iloc` excludes the ending position.
4. **Passing column names to `.iloc`:** Use integer column positions with `.iloc`.
5. **Using parentheses instead of square brackets:** Use `df.loc[...]`, not `df.loc(...)`.
6. **Assuming a selected Series is always a view:** Use `.copy()` when you need an independent object.
7. **Confusing index labels with row positions:** After filtering or resetting an index, labels and positions may differ.

# Pandas: Filtering Data

## 1. What Is Data Filtering?

**Data filtering** is the process of selecting only those rows from a DataFrame that satisfy specific conditions.

In real-world projects, datasets may contain thousands or millions of records. We rarely need every row for every task.

For example, in an Air Quality Index (AQI) dataset, we may want to:
- Find cities where AQI is greater than 100.
- Identify cities where AQI is greater than 100 and temperature exceeds 30°C.
- Retrieve only the city, AQI, and temperature columns.
- Find records where humidity exceeds a specified threshold.

Pandas supports these operations through **Boolean filtering** and the **`query()` method**.

## 2. Boolean Filtering — Using Conditions

Boolean filtering uses a condition that produces `True` or `False` for each row.

Consider:

```python
df_aqi["aqi"] > 100
```

This expression checks each value in the `aqi` column and returns a Boolean Series.

- `True` means the row satisfies the condition.
- `False` means the row does not satisfy the condition.

To retrieve only the matching rows, use the Boolean Series inside the DataFrame's square brackets.

```python
df_aqi[df_aqi["aqi"] > 100]
```

This returns the rows where AQI is greater than 100.

**Important:** `df_aqi[["aqi"] > 100]` is incorrect because `["aqi"]` is a Python list, not the AQI column's values. Use `df_aqi["aqi"]` to access the Series.

## 3. Filtering with Multiple Conditions

Often, a single condition is not enough. We may need to combine two or more conditions.

### A. AND — `&`

The `&` operator requires both conditions to be true.

For example, to select cities where AQI is greater than 100 **and** temperature is greater than 30°C:

```python
df_aqi[
    (df_aqi["aqi"] > 100) &
    (df_aqi["temperature"] > 30)
]
```

A row is selected only if both conditions are satisfied.

### B. OR — `|`

The `|` operator selects a row if at least one condition is true.

For example, select cities where AQI is greater than 150 or temperature exceeds 35°C:

```python
df_aqi[
    (df_aqi["aqi"] > 150) |
    (df_aqi["temperature"] > 35)
]
```

### C. NOT — `~`

The `~` operator negates a Boolean condition.

For example, select rows where AQI is not greater than 100:

```python
df_aqi[~(df_aqi["aqi"] > 100)]
```

For ordinary numerical values, this selects rows where AQI is less than or equal to 100. Missing values require additional care because comparisons involving missing data can produce missing Boolean results.

### Why Do We Use Parentheses?

Each condition should be placed inside parentheses when using `&` or `|`.

Correct:

```python
df_aqi[(df_aqi["aqi"] > 100) & (df_aqi["temperature"] > 30)]
```

This makes the intended comparison order explicit and avoids operator-precedence errors.

**Remember:** In Pandas Boolean filtering, use `&`, `|`, and `~` instead of Python's `and`, `or`, and `not` for element-wise conditions on Series.

## 4. Comparison Operators

These operators define which values should be selected.

| Operator | Meaning | Example |
|---|---|---|
| `>` | Greater than | `df_aqi["aqi"] > 100` |
| `<` | Less than | `df_aqi["temperature"] < 30` |
| `>=` | Greater than or equal to | `df_aqi["humidity"] >= 80` |
| `<=` | Less than or equal to | `df_aqi["aqi"] <= 100` |
| `==` | Equal to | `df_aqi["country"] == "India"` |
| `!=` | Not equal to | `df_aqi["city"] != "Delhi"` |

**Important:** Use `==` to check equality. A single `=` is used for assignment in Python, not for comparison.

## 5. Filtering Rows and Selecting Columns Together

Sometimes we need only specific columns from the filtered results.

For example, select cities where AQI exceeds 100 and temperature exceeds 30°C, but display only the city, AQI, and temperature columns:

```python
df_aqi[
    (df_aqi["aqi"] > 100) &
    (df_aqi["temperature"] > 30)
][["city", "aqi", "temperature"]]
```

The operation has two stages:

1. Filter the rows using the conditions.
2. Select the required columns from those rows.

This is useful when preparing a smaller dataset for reporting, visualisation, or further analysis.

## 6. Understanding Index Labels After Filtering

Filtering usually preserves the original index labels. Pandas does not automatically renumber the filtered rows from zero.

Suppose a filtered DataFrame contains rows with labels `6`, `9`, and `12`.

```python
aqi_data.iloc[0]
```

This retrieves the first row by position, whose index label is `6`.

However:

```python
aqi_data.loc[0]
```

tries to retrieve the row with label `0`. If that label does not exist, Pandas raises a `KeyError`.

To retrieve the row labelled `6`, use:

```python
aqi_data.loc[6]
```

### `.loc` vs `.iloc` After Filtering

| Expression | What it means |
|---|---|
| `aqi_data.loc[6]` | Retrieve the row whose index label is `6` |
| `aqi_data.iloc[0]` | Retrieve the first row by position |
| `aqi_data.loc[6, "city"]` | Retrieve the city value at label `6` |
| `aqi_data.iloc[0, 0]` | Retrieve the first row's first column value |

If you want to reset the index after filtering, you can use `reset_index(drop=True)`. This creates a new sequential index and discards the old index labels.

## 7. Filtering with the `query()` Method

Pandas provides the `query()` method as an alternative to Boolean filtering.

**Syntax:**

`df.query("condition")`

The filtering expression is passed as a string.

For example:

```python
df_aqi.query("aqi > 100")
```

This returns rows where AQI is greater than 100.

For multiple conditions:

```python
df_aqi.query("aqi > 100 and temperature > 30")
```

Inside `query()`, you can use `and` and `or` for logical conditions. The operators `&` and `|` are also supported.

You can select columns after filtering:

```python
df_aqi.query("aqi > 100 and temperature > 30")[
    ["city", "aqi", "temperature"]
]
```

### Boolean Filtering vs `query()`

| Feature | Boolean filtering | `query()` |
|---|---|---|
| Example | `df[df["aqi"] > 100]` | `df.query("aqi > 100")` |
| Condition format | Python expressions | String expression |
| Multiple conditions | Commonly uses `&` and `|` | Supports `and`, `or`, `&`, and `|` |
| Column references | Usually `df["column"]` | Usually column names inside the string |
| Best suited for | Flexible Python-based filtering | Readable, concise filtering expressions |

Both approaches are useful. Choose the one that makes your filtering logic easiest to understand.

## 8. Special Column Names in `query()`

When column names contain spaces or special characters, you may need to surround the column name with backticks inside the query string.

Suppose a DataFrame contains a column named `air quality`.

A query can be written as:

```python
df.query("`air quality` > 100")
```

The backticks tell Pandas to interpret `air quality` as a column name, even though it contains a space.

For ordinary column names such as `aqi`, `temperature`, and `humidity`, backticks are unnecessary.

## 9. Using Python Variables Inside `query()` with `@`

You can use an existing Python variable as a threshold or filtering value by prefixing its name with `@` inside the query string.

For example:

```python
aqi_humidity = 80

df_aqi.query("humidity > @aqi_humidity")
```

Here:
- `aqi_humidity` is a Python variable containing the value `80`.
- `@aqi_humidity` tells `query()` to use that Python variable's value.
- The result contains rows where humidity is greater than 80.

Without `@`, Pandas generally interprets the name as a column or another name in the query expression.

This is particularly useful when thresholds change dynamically, such as when a user chooses a value in a dashboard.

## 10. Chained Comparisons in `query()`

The `query()` method supports chained comparisons, which can make range filtering more readable.

For example:

```python
df_aqi.query("100 < aqi < 200")
```

This selects rows where AQI is strictly greater than 100 and strictly less than 200.

It is equivalent in meaning to:

```python
df_aqi[(df_aqi["aqi"] > 100) & (df_aqi["aqi"] < 200)]
```

Because the bounds are strict, values equal to 100 or 200 are excluded.

For inclusive bounds, you can use:

```python
df_aqi.query("100 <= aqi <= 200")
```

This includes both boundary values.

## 11. Does `query()` Always Return a Copy?

By default, `query()` returns a filtered DataFrame rather than modifying the original DataFrame in place.

However, do not assume that every result is necessarily a fully independent, deep copy of the underlying data. The details of data sharing and mutation depend on the operation and Pandas' Copy-on-Write behaviour.

If you plan to modify the filtered result independently, explicitly use `.copy()`:

```python
aqi_data = df_aqi.query("aqi > 100").copy()
```

This makes your intention clear and is a good habit when creating a separate working dataset.

# Pandas Data Cleaning — Part 1: Handling Missing Values

## 1. What Is Data Cleaning?

**Data cleaning** is the process of identifying and correcting problems in a dataset so that the data becomes more reliable and suitable for analysis.

Real-world datasets often contain:
- Missing values.
- Duplicate records.
- Incorrect data types.
- Inconsistent formats.
- Invalid or unexpected values.

In this section, we focus on handling **missing values**.

### Why Are Values Missing?

Missing data can occur for several reasons:
- A user did not provide their age during registration.
- A sensor failed to record a measurement.
- A survey respondent skipped a question.
- Data was lost during collection or transfer.
- Different data sources contain different information.

Missing values must be handled carefully because they can affect calculations, visualisations, and machine learning models.

## 2. What Is NaN?

`NaN` stands for **Not a Number**. In Pandas, it is commonly used to represent missing numerical data.

For example:

| Name | Age |
|---|---:|
| Priyam | 21 |
| Palak | NaN |
| Rupak | 24 |

Here, Palak's age is missing. It does not mean her age is zero; it means the value is unknown or unavailable.

Depending on the data and its type, Pandas may also represent missing values using `None`, `NaT`, or the newer `pd.NA` marker.

**Important:** Missing values and zero are not the same. Replacing missing values with zero is appropriate only when zero has a meaningful interpretation for that particular column.

## 3. Detecting Missing Values: `isnull()` and `isna()`

Pandas provides two equivalent methods to identify missing values:

- `df.isnull()`
- `df.isna()`

Both return a DataFrame of Boolean values.

- `True` means the value is missing.
- `False` means the value is present.

### Conceptual Example

| Name | Age | Salary |
|---|---:|---:|
| Priyam | 21 | 45000 |
| Palak | NaN | 38000 |
| Rupak | 24 | NaN |

The missing-value check identifies Palak's missing age and Rupak's missing salary.

**Key point:** `isnull()` and `isna()` are aliases, meaning they perform the same operation.

## 4. Counting Missing Values with `sum()`

To count missing values in each column, combine `isnull()` or `isna()` with `sum()`.

Conceptually:

`df.isnull().sum()`

This works because Boolean `True` values are treated as `1` and `False` values as `0` when summed.

For example:

| Column | Missing Count |
|---|---:|
| Name | 0 |
| Age | 1 |
| Salary | 1 |

This tells us which columns contain missing values and how many are present.

### Count Missing Values in the Entire DataFrame

`df.isnull().sum().sum()`

The first `sum()` counts missing values in each column. The second adds those column-wise counts to get the total number of missing values in the DataFrame.

**Real-world use:** Before cleaning a dataset, count missing values to decide which columns need attention.

## 5. Removing Missing Values with `dropna()`

The `dropna()` method removes rows or columns containing missing values.

### A. Drop Rows

`df.dropna()`

By default, `dropna()` removes every row containing at least one missing value.

For example, if a customer record has a missing age, the entire row may be removed—even if the customer's name and email are available.

### B. Drop Columns

`df.dropna(axis=1)`

Setting `axis=1` removes columns containing missing values instead of rows.

The default is `axis=0`, which means rows.

### Important Parameters

| Parameter | Purpose |
|---|---|
| `axis=0` | Remove rows |
| `axis=1` | Remove columns |
| `how="any"` | Drop a row or column if any value is missing; this is the default |
| `how="all"` | Drop only if all values are missing |
| `subset` | Consider missing values only in specified columns |
| `inplace` | Modify the original object when supported and set to `True` |

### Why Should We Be Careful with `dropna()`?

Removing rows may cause useful information to be lost. Removing columns may discard important features.

For example, if only 2 out of 10,000 employee records have a missing age, dropping those rows may be reasonable. But if 40% of the ages are missing, deleting all those rows may significantly reduce the dataset.

**Best practice:** Understand why values are missing and how much data would be lost before dropping anything.

Also, `dropna()` returns a new object by default; it does not automatically change the original DataFrame.

## 6. Replacing Missing Values with `fillna()`

The `fillna()` method replaces missing values with a specified value or an appropriate filling strategy.

### A. Fill with a Constant

Conceptually:

`df.fillna(0)`

This replaces missing values throughout the DataFrame with zero.

Other possible replacement values include:
- `"Unknown"` for an appropriate categorical column.
- A known default value when it is meaningful.
- A calculated statistic, such as the mean or median.

**Important:** One replacement value may not be suitable for every column. For example, zero might be appropriate for a missing count in some datasets but misleading for a missing salary.

### B. Fill a Specific Column

You can fill missing values in only one column instead of the entire DataFrame.

For example, replacing missing ages with a calculated age statistic avoids unintentionally replacing missing values in unrelated columns.

## 7. Mean Imputation

**Mean imputation** means replacing missing numerical values with the mean of the available values in that column.

The mean is calculated as:

\[
\text{Mean}=\frac{\text{Sum of observed values}}{\text{Number of observed values}}
\]

For example, consider the ages:

`20, 30, NaN, 40`

The mean of the observed values is:

\[
\frac{20+30+40}{3}=30
\]

After mean imputation, the age values become:

`20, 30, 30, 40`

### Why Use Mean Imputation?

- It retains rows that would otherwise contain missing numerical values.
- It is easy to understand and implement.
- It can be useful when a simple baseline cleaning method is sufficient.

### Limitations of Mean Imputation

- It can reduce the natural variability of the data.
- It may distort relationships between variables.
- It is sensitive to extreme values, also called outliers.
- It is not suitable for every type of column.

For skewed numerical data or data containing outliers, median imputation may be more appropriate.

**Important:** Calculate the mean using observed values, not by treating missing values as zero. Pandas' `mean()` normally excludes missing values.

For machine learning, calculate imputation statistics using the training data only, then apply those learned values to validation and test data. This prevents data leakage.

## 8. Why Create a Copy Before Cleaning?

Suppose `df_raw` contains the original dataset. You want to clean it while keeping the original data available for comparison or recovery.

A common approach is:

`cleaned_data = df_raw.copy()`

This creates an independent copy of the DataFrame.

You can then apply cleaning operations to `cleaned_data` without intentionally modifying `df_raw`.

For example, after calculating the mean age, fill missing values in the copied DataFrame's age column and assign the result back to that column.

### Why Does Assignment Matter?

An expression such as:

`df_raw["age"].fillna(age_mean)`

returns a Series with missing ages filled, but it does not automatically assign that result back to the original DataFrame.

To retain the cleaned values, assign the result to the relevant column or use an appropriate supported in-place operation.

**Best practice:** Keep the raw data unchanged and perform cleaning on a separate working DataFrame. This makes it easier to compare the original and cleaned data.

## 9. Forward Fill: `ffill()`

Forward fill replaces a missing value with the most recent preceding non-missing value in the selected direction.

Conceptual example:

| Row | Temperature |
|---:|---:|
| 0 | 25 |
| 1 | NaN |
| 2 | NaN |
| 3 | 28 |

After forward fill:

| Row | Temperature |
|---:|---:|
| 0 | 25 |
| 1 | 25 |
| 2 | 25 |
| 3 | 28 |

The missing values inherit the last observed value, `25`.

### When Is Forward Fill Useful?

- Time-series data.
- Sensor readings recorded at regular intervals.
- Records where the previous known value remains valid until updated.

**Caution:** Forward fill is not appropriate when the previous value cannot reasonably represent the missing observation.

By default, `ffill()` operates down rows (`axis=0`). The original DataFrame remains unchanged unless the result is assigned back or an applicable in-place option is used.

## 10. Backward Fill: `bfill()`

Backward fill replaces a missing value with the next available non-missing value in the selected direction.

Using the previous example:

| Row | Temperature |
|---:|---:|
| 0 | 25 |
| 1 | NaN |
| 2 | NaN |
| 3 | 28 |

After backward fill:

| Row | Temperature |
|---:|---:|
| 0 | 25 |
| 1 | 28 |
| 2 | 28 |
| 3 | 28 |

The missing values inherit the next observed value, `28`.

### When Is Backward Fill Useful?

It can be useful when the next known observation is a reasonable replacement for the missing value.

**Caution:** Backward fill uses future observations relative to the missing row. In forecasting or time-sensitive machine learning tasks, this may introduce data leakage if those future values would not have been available at prediction time.

## 11. `ffill()` vs `bfill()`

| Feature | Forward Fill (`ffill`) | Backward Fill (`bfill`) |
|---|---|---|
| Replacement source | Previous non-missing value | Next non-missing value |
| Direction | Forward through the rows | Backward through the rows |
| Useful for | Carrying the last known value forward | Using the next known value |
| Main risk | Previous value may no longer be valid | Future information may be unavailable or cause leakage |

Both methods depend on row order. For time-series data, sort the records chronologically before filling.

Leading missing values may remain after forward fill, and trailing missing values may remain after backward fill if there is no available value in the required direction.

## 12. Comparing Missing-Value Handling Methods

| Method | What it does | When to consider it |
|---|---|---|
| `isnull()` / `isna()` | Identifies missing values | Detecting missing data |
| `isnull().sum()` | Counts missing values per column | Assessing data quality |
| `dropna()` | Removes rows or columns with missing values | When removal is justified |
| `fillna(value)` | Replaces missing values with a specified value | When a meaningful replacement exists |
| Mean imputation | Fills numerical gaps with the observed mean | Simple numerical imputation |
| `ffill()` | Uses the preceding observed value | Suitable sequential data |
| `bfill()` | Uses the next observed value | Suitable cases where later values are valid replacements |
| `.copy()` | Creates a separate DataFrame copy | Preserving raw data during cleaning |

## 13. A Practical Missing-Value Workflow

A sensible workflow for a new dataset is:

1. Load the raw dataset.
2. Inspect its structure and data types.
3. Identify missing values using `isna()`.
4. Count missing values with `isna().sum()`.
5. Understand why the values are missing.
6. Choose an appropriate strategy for each column.
7. Preserve the original dataset and clean a copy.
8. Verify that the cleaning worked by checking missing-value counts again.

The goal is not simply to eliminate every missing value. The goal is to make the data reliable without introducing unnecessary bias or losing valuable information.

## 14. Key Takeaways

- `NaN` represents missing data, not necessarily zero.
- `isnull()` and `isna()` perform the same missing-value check.
- `isnull().sum()` counts missing values in each column.
- `dropna()` removes rows by default; `axis=1` removes columns.
- `fillna()` replaces missing values, but the replacement should make sense for the data.
- Mean imputation uses the mean of observed numerical values and may not be suitable for skewed data.
- `ffill()` uses the previous observed value, while `bfill()` uses the next one.
- Cleaning operations often return a new object; assign the result if you want to retain it.
- Using `.copy()` helps preserve the original dataset.
- Always validate the cleaned dataset after applying a strategy.

# Handling Duplicate Values in Pandas

## 1. What Are Duplicate Values?

Duplicate values are repeated records in a dataset.

For example, suppose an employee dataset contains the same employee information more than once. These repeated records are called duplicate rows.

Duplicates can cause problems such as:
- Incorrect calculations and statistics.
- Inflated employee or customer counts.
- Inaccurate data analysis.
- Unreliable business decisions.

Pandas provides methods to identify and remove duplicate records.

## 2. Creating a Sample DataFrame

```python
import pandas as pd

employee_data = {
    "name": ["Aman", "Priya", "Rahul", "Aman", "Neha", "Priya"],
    "age": [24, 22, 25, 24, 23, 22],
    "department": ["IT", "HR", "Finance", "IT", "HR", "HR"]
}

df_employee = pd.DataFrame(employee_data)

df_employee
```

The DataFrame contains six rows. However, some records are repeated.

- Aman appears twice with identical information.
- Priya appears twice with identical information.

Therefore, there are two duplicate rows.

## 3. Identifying Duplicate Rows — `duplicated()`

The `duplicated()` method identifies duplicate rows in a DataFrame.

**Syntax:**

```python
df.duplicated()
```

Example:

```python
df_employee.duplicated()
```

Expected output:

```text
0    False
1    False
2    False
3     True
4    False
5     True
dtype: bool
```

Explanation:

- `False`: This is the first occurrence of a record.
- `True`: This record has already appeared earlier.

By default, Pandas considers the first occurrence unique and marks subsequent occurrences as duplicates.

**Important:** `duplicated()` does not delete any rows. It only identifies duplicate records.

## 4. Understanding the `keep` Parameter

The `duplicated()` method provides a `keep` parameter to control which occurrence should be considered unique.

### A. `keep="first"` (Default)

```python
df_employee.duplicated(keep="first")
```

- Keeps the first occurrence as unique.
- Marks all subsequent occurrences as duplicates.
- This is the default behaviour.

### B. `keep="last"`

```python
df_employee.duplicated(keep="last")
```

- Keeps the last occurrence as unique.
- Marks previous occurrences as duplicates.

### C. `keep=False`

```python
df_employee.duplicated(keep=False)
```

- Marks every occurrence of a duplicated record as `True`.
- Even the first occurrence is marked as duplicate.
- Useful when we want to identify all rows belonging to duplicate groups.

**Remember:**

| Parameter | Behaviour |
|---|---|
| `keep="first"` | Keeps the first occurrence |
| `keep="last"` | Keeps the last occurrence |
| `keep=False` | Marks all duplicate occurrences as True |

## 5. Finding Duplicates in a Specific Column

Sometimes we don't want to check the entire row. We only want to identify repeated values in a particular column.

Example:

```python
df_employee["name"].duplicated()
```

This checks whether employee names have appeared previously.

We can also use:

```python
df_employee["name"].duplicated(keep=False)
```

This marks all occurrences of repeated names.

**Important difference:**

```python
df_employee.duplicated()
```

Checks the complete row.

```python
df_employee["name"].duplicated()
```

Checks only the `name` column.

Two employees can have the same name but different ages or departments. In that case, they are duplicates in the name column, but not necessarily duplicate rows.

## 6. Counting Duplicate Rows

To count duplicate rows, we can combine `duplicated()` with `sum()`.

```python
df_employee.duplicated().sum()
```

Expected output:

```text
2
```

**Why does this work?**

- `duplicated()` returns Boolean values.
- In Python, `True` is treated as `1`.
- `False` is treated as `0`.
- Therefore, `sum()` counts the number of `True` values.

This is a useful way to calculate the number of duplicate rows.

## 7. Removing Duplicate Rows — `drop_duplicates()`

The `drop_duplicates()` method removes duplicate rows from a DataFrame.

**Syntax:**

```python
df.drop_duplicates()
```

Example:

```python
df_employee.drop_duplicates()
```

This returns a DataFrame containing only unique rows.

By default:
- The first occurrence is retained.
- Subsequent duplicate occurrences are removed.
- The original DataFrame is not modified unless we assign the result or use `inplace=True`.

## 8. Understanding `inplace=True`

The `inplace` parameter controls whether an operation modifies the original DataFrame.

### Without `inplace=True`

```python
df_cleaned = df_employee.drop_duplicates()
```

- Creates a cleaned result.
- Stores it in `df_cleaned`.
- Original `df_employee` remains unchanged.

### With `inplace=True`

```python
df_employee.drop_duplicates(inplace=True)
```

- Directly modifies the original DataFrame.
- Duplicate rows are removed from `df_employee`.
- No new cleaned DataFrame is returned.

**Important:** When `inplace=True` is used, the method returns `None`, so avoid assigning its result back to the DataFrame.

## 9. Why Use `.copy()` Before Cleaning?

Sometimes we want to preserve the original dataset while performing cleaning operations.

For this, we use `.copy()`.

Example:

```python
df_cleaned = df_employee.copy()

df_cleaned.drop_duplicates(inplace=True)
```

Now:

- `df_employee` contains the original data.
- `df_cleaned` contains the cleaned data.

This is useful when we want to compare the original and cleaned datasets or preserve the raw data for future reference.

## 10. Complete Example

```python
import pandas as pd

employee_data = {
    "name": ["Aman", "Priya", "Rahul", "Aman", "Neha", "Priya"],
    "age": [24, 22, 25, 24, 23, 22],
    "department": ["IT", "HR", "Finance", "IT", "HR", "HR"]
}

df_employee = pd.DataFrame(employee_data)

# Identify duplicate rows
df_employee.duplicated()

# Identify all occurrences of duplicate rows
df_employee.duplicated(keep=False)

# Find duplicates in a specific column
df_employee["name"].duplicated()

# Count duplicate rows
df_employee.duplicated().sum()

# Create a copy of the original DataFrame
df_cleaned = df_employee.copy()

# Remove duplicate rows from the copied DataFrame
df_cleaned.drop_duplicates(inplace=True)

# Display cleaned data
df_cleaned
```

## 11. Important Points to Remember

1. `duplicated()` identifies duplicate rows but does not remove them.
2. `drop_duplicates()` removes duplicate rows.
3. By default, the first occurrence is retained.
4. `keep="last"` retains the last occurrence.
5. `keep=False` marks all occurrences of duplicate records.
6. `df["column"].duplicated()` checks duplicates in a particular column.
7. `duplicated().sum()` counts duplicate rows.
8. `inplace=True` modifies the original DataFrame.
9. `.copy()` helps preserve the original dataset before cleaning.
10. Duplicate names do not necessarily mean duplicate records. Always consider the columns that define a genuinely repeated record.

## Quick Revision

| Method / Parameter | Purpose |
|---|---|
| `duplicated()` | Identify duplicate rows |
| `duplicated(keep="first")` | Keep first occurrence |
| `duplicated(keep="last")` | Keep last occurrence |
| `duplicated(keep=False)` | Mark all duplicate occurrences |
| `df["name"].duplicated()` | Check duplicates in one column |

# Handling Data Types in Pandas

## 1. What Are Data Types?

A data type defines what kind of value a variable or column stores.

For example:

- `21` is an integer.
- `85.5` is a floating-point number.
- `"Aman"` is a string.
- `True` is a Boolean value.

In Pandas, every DataFrame column has a data type.

Data types are important because they determine how Pandas stores, processes, and performs operations on data.

## 2. Common Data Types

| Data Type | Description | Example |
|---|---|---|
| `int64` | Integer values | 10, 25, 100 |
| `float64` | Decimal values | 10.5, 25.75 |
| `object` | General Python objects, often strings | "Aman", "Delhi" |
| `bool` | Boolean values | True, False |
| `datetime64[ns]` | Date and time values | 2026-11-05 |
| `Int64` | Pandas nullable integer type | 10, 20, `<NA>` |

**Note:** `int64` and `Int64` are different. The capital-I `Int64` is a Pandas nullable integer extension type.

## 3. Creating a Sample DataFrame


- `name`: String values.
- `age`: Integer values.
- `marks`: Decimal values.
- `passed`: Boolean values.

Sometimes data is stored in an incorrect or unsuitable data type.

- A numerical column may be stored as a string.

### Important: `astype("int64")` Does Not Round Decimal Values

```python

```

This rounds the values before converting them to integers.

**Important:** Converting data types can change or lose information. Always check whether the conversion is appropriate for the data.

## 8. Understanding Missing Values and Integer Data Types

Consider this DataFrame:


Traditional NumPy integer types such as `int64` cannot directly represent a missing value like `NaN`.

Therefore, when a numerical column contains a missing value, Pandas commonly converts it into a floating-point type.

For example:

```text

22
NaN

```

The column becomes `float64`.

This allows Pandas to represent both numbers and `NaN`.

**Remember:** The capital `I` in `Int64` is important.

- `Int64`: Pandas nullable integer type.

# Handling Date-Time in Pandas

## 1. What is Date-Time?

Date-Time represents the date and time at which a particular event occurs.

Examples:

- `2026-11-05` → Date
- `10:30:45` → Time
- `2026-11-05 10:30:45` → Date and Time

### Real-World Applications

- Recording online order dates.
- Tracking bank transactions.
- Managing employee attendance.
- Analyzing daily weather and air quality.
- Tracking delivery dates and durations.

## 2. Why Do We Need Date-Time Handling?

In real-world datasets, dates are often stored as strings rather than actual datetime values.

For example, an online shopping dataset may contain order dates.

| Order ID | Order Date |
|---|---|
| 101 | 2026-11-01 |
| 102 | 2026-11-02 |
| 103 | 2026-11-03 |

We may need to:

- Filter records based on a specific date.
- Find the number of orders in a particular month.
- Sort records from oldest to newest.
- Extract year, month and day.
- Calculate the difference between two dates.

To perform these operations efficiently, we convert date strings into proper datetime values.

## 3. String vs Date-Time

A date written inside quotation marks is generally treated as a string.

```python
date_str = "2026-12-31"

type(date_str)
```

Output:

```text
str
```

A string represents text. Python does not automatically treat every string that looks like a date as an actual date.

Pandas provides `pd.to_datetime()` to convert date representations into datetime values.

## 4. pd.to_datetime()

`pd.to_datetime()` is a Pandas function used to convert date and time representations into proper datetime values.

### Syntax

```python
pd.to_datetime(value)
```

### Example

```python
date_str = "2026-12-31"

date = pd.to_datetime(date_str)

type(date)
```

Output:

```text
pandas.Timestamp
```

After conversion, Pandas understands the value as a date-time rather than ordinary text.

## 5. What is Timestamp?

`Timestamp` is a Pandas date-time object used to represent a specific date and time.

Example:

```python
pd.to_datetime("2026-12-31 10:30:00")
```

A Timestamp can represent:

- Year
- Month
- Day
- Hour
- Minute
- Second

For example:

`2026-12-31 10:30:00`

- Year = 2026
- Month = 12
- Day = 31
- Hour = 10
- Minute = 30
- Second = 00

## 6. Converting a DataFrame Column into Date-Time

Sometimes dates are stored in a DataFrame column as strings.

We can convert the entire column using `pd.to_datetime()`.

### Example

```python
df["date"] = pd.to_datetime(df["date"])
```

After conversion, the column generally has the following dtype:

```text
datetime64[ns]
```

### Important Difference

| Type | Meaning |
|---|---|
| `str` / `object` | Date stored as text |
| `Timestamp` | Individual Pandas date-time value |
| `datetime64[ns]` | Pandas datetime column |
| `NaT` | Missing or invalid datetime value |

## 7. Different Date Formats

Dates can appear in different formats in a dataset.

Examples:

- `2026-11-05`
- `05-11-2026`
- `11/05/2026`
- `05 Nov 2026`

Pandas can parse many common date formats automatically.

However, when the format is ambiguous or needs to be explicitly specified, we can use the `format` parameter.

### Example

```python
pd.to_datetime("05-11-2026", format="%d-%m-%Y")
```

### Common Format Codes

| Format Code | Meaning |
|---|---|
| `%d` | Day |
| `%m` | Month |
| `%Y` | Four-digit year |
| `%H` | Hour (24-hour format) |
| `%M` | Minute |
| `%S` | Second |

**Note:** Always understand the date format before conversion. For example, `05-11-2026` can be ambiguous if the day-month order is unknown.

## 8. Handling Invalid Dates

Sometimes datasets contain invalid date values.

Example:

```text
2026-11-05
2026-11-06
Hello
2026-11-08
```

The value `Hello` cannot be interpreted as a valid date.

Pandas provides the `errors` parameter.

### errors="raise"

This is the default behavior. Pandas raises an error when it encounters an invalid date.

```python
pd.to_datetime(df["date"], errors="raise")
```

### errors="coerce"

Invalid date values are converted into `NaT`.

```python
pd.to_datetime(df["date"], errors="coerce")
```

### What is NaT?

`NaT` stands for **Not a Time**.

It represents missing or invalid datetime values.

| Data Type | Missing Value |
|---|---|
| Numerical | `NaN` |
| Datetime | `NaT` |

## 9. Extracting Date Components Using .dt

Pandas provides the `.dt` accessor to work with datetime Series.

It allows us to extract different components from a date-time column.

**Important:** `.dt` is used with datetime Series. For an individual Pandas Timestamp, we can directly access attributes such as `.year`, `.month` and `.day`.

# Sorting Date-Time Data in Pandas

## 1. Why Sort Date-Time Data?

Sorting arranges records in chronological order.

It is useful when analysing:

- Oldest and newest transactions.
- First and last orders.
- Daily activity.
- Historical records.
- Time-based trends.

## 2. Sorting in Ascending Order

Use `sort_values()` to sort a DataFrame based on a date column.

**Syntax:**

```python
df.sort_values(by="date")
```

Ascending order is the default.

It arranges dates from oldest to newest.

## 3. Sorting in Descending Order

Use `ascending=False` to arrange dates from newest to oldest.

```python
df.sort_values(by="date", ascending=False)
```

## 4. Selecting Specific Columns After Sorting

We can select particular columns after sorting.

```python
df.sort_values(by="order_date")[
    ["order_id", "order_date"]
]
```

This sorts the DataFrame and displays only the selected columns.

## 5. Understanding Index Preservation

By default, `sort_values()` preserves the original index labels.

For example, after sorting, the index may appear as:

`3, 1, 4, 0, 2`

This is normal. The rows have been rearranged, but their original index labels remain.

## 6. Resetting the Index

Use `reset_index()` to create a new sequential index.

### `reset_index()`

```python
df.reset_index()
```

By default, the old index is preserved as a new column.

### `reset_index(drop=True)`

```python
df.reset_index(drop=True)
```

The old index is discarded, and a fresh index is created starting from zero.

## 7. Using `inplace=True`

We can modify the original DataFrame directly:

```python
df.sort_values(by="date", inplace=True)
```

When `inplace=True` is used, the operation modifies the existing DataFrame and returns `None`.

Therefore, avoid assigning the result back to the same variable:

```python
df = df.sort_values(by="date", inplace=True)
```

This would make `df` equal to `None`.

| `ascending=True` | Oldest to newest for dates |

| `reset_index()` | Reset index while retaining the old index as a column |
| `reset_index(drop=True)` | Reset index and discard the old index |
| `inplace=True` | Modify the original DataFrame |

**Key takeaway:** Use `sort_values()` for chronological ordering and `reset_index(drop=True)` when you want a clean sequential index after sorting.

Before using `.dt`, make sure the column has been converted into a Date-Time dtype.

```python
df["date"] = pd.to_datetime(df["date"])
```

## 2. Extracting Date Components

```python
df["year"] = df["date"].dt.year
```

```python
df["month_name"] = df["date"].dt.month_name()
```

### `.dt.day_name()`

Returns the weekday name.

```python
df["day_name"] = df["date"].dt.day_name()
```

Example output: `Thursday`

## 4. `.dt` vs Direct Attributes

The syntax depends on whether we are working with a Series or a single Timestamp.

**For a DataFrame column (Series):**

```python
df["date"].dt.year
```

**For a single Timestamp:**

```python
date.year
date.month
```

A single Timestamp does not require `.dt`.

## 5. Practical Applications

# Date Difference and Timedelta in Pandas

## 1. What is Date Difference?

Date difference means calculating the amount of time between two dates or timestamps.

For example:

- Order Date: `2026-11-01`
- Delivery Date: `2026-11-05`
- Delivery Duration: 4 days

Pandas automatically calculates the difference when one Date-Time value is subtracted from another.

## 2. Calculating Date Difference

**Syntax:**

```python
df["duration"] = df["end_date"] - df["start_date"]
```

Example:

```python
df["delivery_time"] = (
    df["delivery_date"] - df["order_date"]
)
```

The result is a Timedelta value.

## 3. What is Timedelta?

`Timedelta` represents a duration or difference between two points in time.

Examples:

- 3 days
- 5 hours
- 30 minutes
- 45 seconds

When we subtract two Date-Time values, Pandas returns a Timedelta Series.

Its dtype is generally displayed as `timedelta64` with a particular resolution.

## 4. Extracting the Number of Days

Use `.dt.days` to extract the whole-day component from a Timedelta Series.

```python
df["delivery_days"] = df["delivery_time"].dt.days
```

The extracted day values are generally stored as integers.

**Important:** `.dt.days` extracts the day component, not a rounded-up total number of days. For example, a duration of 2 days and 12 hours gives `2`.

## 5. Creating a Timedelta Manually

Pandas provides `pd.Timedelta()` to create a duration.

**Syntax:**

```python
pd.Timedelta(days=3)
```

Other supported units include:

```python
pd.Timedelta(hours=5)
pd.Timedelta(minutes=30)
pd.Timedelta(seconds=45)
```

## 6. Comparing Durations

Timedelta values can be compared using normal comparison operators.

Example:

```python
df["fast_delivery"] = (
    df["delivery_time"] <= pd.Timedelta(days=3)
)
```

This creates a Boolean column:

- `True` → Delivery took 3 days or less.
- `False` → Delivery took more than 3 days.

Remember the difference:

- `<` means strictly less than.
- `<=` means less than or equal to.

## 7. Practical Applications

Timedelta is useful for:

- Calculating delivery duration.
- Measuring employee working hours.
- Finding response times.
- Calculating time between transactions.
- Checking whether a task was completed within a deadline.

**Key takeaway:** Subtract Date-Time values to calculate duration, use `.dt.days` to extract whole days, and use `pd.Timedelta()` to create or compare durations.

````python
data = {
    "order_id": [101, 102, 103, 104],
    "order_date": [
        "2026-11-01",
        "2026-11-03",
        "2026-11-05",
        "2026-11-07"
    ],
    "delivery_date": [
        "2026-11-04",
        "2026-11-08",
        "2026-11-06",
        "2026-11-12"
    ]
}

df_order = pd.DataFrame(data)

#converting both order date and delivery date into date time format
df_order["order_date"] = pd.to_datetime(df_order["order_date"])

df_order["delivery_date"] = pd.to_datetime(df_order["delivery_date"])

#difference of delivery and order date
df_order["delivery_date"] - df_order["order_date"]   # give no of days and its data type is : dtype: timedelta64[us]
(df_order["delivery_date"] - df_order["order_date"]).dt.days  # also gives no fo days but its data type is: dtype: int64

#store delivery time in a new column 
df_order["delivery_time"] = df_order["delivery_date"] - df_order["order_date"]

df_order["delivery_time"].dtype  #dtype('<M8[us]')

#make a new column called fast_delivery and compare delivery timing
df_order["fast_delivery"] = df_order["delivery_time"] <= pd.Timedelta(days=3)
df_order["fast_delivery"]
````

# Date Arithmetic in Pandas

## 1. What is Date Arithmetic?

Date arithmetic means performing mathematical operations on dates.

It allows us to add or subtract a duration from a date.

Examples:

- Calculate an expected delivery date.
- Find a reminder date before an appointment.
- Calculate a subscription expiry date.
- Find a deadline after a given number of days.

## 2. Using `pd.Timedelta()`

Pandas provides `pd.Timedelta()` to represent a duration that can be added to or subtracted from a Date-Time value.

**Syntax:**

```python
pd.Timedelta(days=5)
```

Other supported units include:

```python
pd.Timedelta(hours=3)
pd.Timedelta(minutes=30)
pd.Timedelta(seconds=15)
```

## 3. Adding Days to a Date

Use the `+` operator to add a duration.

```python
df["expected_delivery"] = (
    df["order_date"] + pd.Timedelta(days=5)
)
```

This calculates a date five days after the order date.

## 4. Subtracting Days from a Date

Use the `-` operator to subtract a duration.

```python
df["reminder_date"] = (
    df["order_date"] - pd.Timedelta(days=2)
)
```

This calculates a date two days before the order date.

## 5. Adding or Subtracting Other Units

The same approach works with hours, minutes, and seconds.

```python
df["new_time"] = (
    df["date_time"] + pd.Timedelta(hours=2)
)
```

```python
df["previous_time"] = (
    df["date_time"] - pd.Timedelta(minutes=30)
)
```

## 6. Important Points

- `pd.Timedelta()` represents a duration.
- `+` adds a duration to a date or timestamp.
- `-` subtracts a duration from a date or timestamp.
- The result remains Date-Time data.
- The original date column is not modified unless you assign the result back to it.

## 7. Practical Applications

Date arithmetic is useful for:

- Expected delivery dates.
- Payment deadlines.
- Appointment reminders.
- Subscription expiry dates.
- Project completion dates.
- Scheduling future events.

**Key takeaway:** Use `pd.Timedelta()` with `+` or `-` to perform Date-Time calculations in Pandas.

````python
import pandas as pd

data = {
    "order_id": [101, 102, 103],
    "order_date": [
        "2026-11-01",
        "2026-11-05",
        "2026-11-10"
    ]
}
df_order = pd.DataFrame(data)
df_order["order_date"] = pd.to_datetime(df_order["order_date"])

#make a new column of expected_delivery and add 5days in it from order_date
df_order["expected_delivery"] = df_order["order_date"] + pd.Timedelta(days=5)

#amke a reminder_date
df_order["reminder_date"] = df_order["order_date"] - pd.Timedelta(days=2)

df_order

#ckeck data type
print(df_order["expected_delivery"].dtype)
print(df_order["reminder_date"].dtype)
````

# Resampling in Pandas

## 1. What is Resampling?

Resampling is the process of converting time-series data from one frequency to another.

In simple words, it means grouping date-based data into different time intervals, such as:

- Daily
- Weekly
- Monthly
- Quarterly
- Yearly

For example, an e-commerce company may have daily sales records but wants to calculate:

- Total sales per week.
- Total sales per month.
- Average monthly revenue.
- Total yearly sales.

Pandas provides the `resample()` method to perform these operations.

**Real-world applications:**
- Monthly revenue analysis.
- Weekly website traffic analysis.
- Daily stock market analysis.
- Monthly customer activity analysis.
- Quarterly business performance reports.

## 2. Creating a Sample Dataset

Let's create a sales dataset containing dates from January to December 2026.

```python
import pandas as pd

data = {
    "date": [
        "2026-01-05", "2026-01-15", "2026-01-25",
        "2026-02-03", "2026-02-14",
        "2026-03-05", "2026-03-20",
        "2026-04-10", "2026-04-22",
        "2026-05-01", "2026-05-18",
        "2026-06-07", "2026-06-25",
        "2026-07-04", "2026-07-19",
        "2026-08-08", "2026-08-21",
        "2026-09-02", "2026-09-17",
        "2026-10-05", "2026-10-23",
        "2026-11-01", "2026-11-02", "2026-11-03",
        "2026-11-04", "2026-11-05", "2026-11-15",
        "2026-12-01", "2026-12-02", "2026-12-10",
        "2026-12-20", "2026-12-28"
    ],
    "sales": [
        1200, 1800, 2200,
        1500, 2500,
        1800, 3200,
        2100, 2900,
        1700, 2600,
        3000, 3500,
        2200, 2800,
        3100, 2700,
        1900, 3300,
        2500, 3600,
        1000, 2000, 1500, 3000, 2500, 1800,
        4000, 3500, 2200, 4500, 2800
    ]
}

df_sales = pd.DataFrame(data)
```

## 3. Converting Date Column into Datetime

Before performing resampling, the date column should be in a proper datetime format.

```python
df_sales["date"] = pd.to_datetime(df_sales["date"])
```

Check the data type:

```python
df_sales["date"].dtype
```

Expected output:

```text
datetime64[ns]
```

**Why is this important?**

Resampling is a time-based operation. Pandas needs to understand the date column as actual datetime values rather than ordinary strings.

## 4. Understanding the `resample()` Method

The `resample()` method groups time-series data into specified time intervals.

### General Syntax

```python
df.resample("frequency", on="date_column")["value_column"].sum()
```

### Understanding Each Component

| Component | Purpose |
|---|---|
| `resample()` | Groups data into time intervals |
| `frequency` | Defines the time interval |
| `on` | Specifies the date column |
| `value_column` | Selects the numerical column |
| `sum()` | Calculates the total for each interval |

**Important:** The date column must be datetime-like. When using `on="date_column"`, you do not need to set the date column as the DataFrame index.

## 5. Understanding Resampling Frequencies

Pandas supports different frequencies for time-based aggregation.

| Frequency | Meaning | Example |
|---|---|---|
| `"D"` | Daily | Every day |
| `"W"` | Weekly | Every week |
| `"ME"` | Month End | End of every month |
| `"MS"` | Month Start | Beginning of every month |
| `"QE"` | Quarter End | End of every quarter |
| `"YE"` | Year End | End of every year |

**Note:** Modern Pandas uses `"ME"` for month-end frequency. Older tutorials may use `"M"`.

Similarly, `"QE"` and `"YE"` are modern quarter-end and year-end aliases.

## 6. Monthly Total Sales Using `resample()`

Suppose we want to calculate the total sales for every month from January to December 2026.

```python
monthly_sales = df_sales.resample(
    "ME",
    on="date"
)["sales"].sum()

monthly_sales
```

### Explanation

1. `resample("ME")` groups records by month-end.
2. `on="date"` specifies the date column.
3. `["sales"]` selects the sales column.
4. `.sum()` calculates the total sales for each month.

### Expected Output

| Month | Total Sales |
|---|---:|
| January | 5,200 |
| February | 4,000 |
| March | 5,000 |
| April | 5,000 |
| May | 4,300 |
| June | 6,500 |
| July | 5,000 |
| August | 5,800 |
| September | 5,200 |
| October | 6,100 |
| November | 11,800 |
| December | 17,000 |

The result contains one aggregated sales value for each month.

## 7. Filtering a Specific Date Range Before Resampling

Sometimes we do not want to perform aggregation on the entire dataset.

For example, suppose we only want data between November and December 2026.

First, filter the required date range:

```python
df_sales[
    (df_sales["date"] >= "2026-11-01") &
    (df_sales["date"] <= "2026-12-31")
]
```

Here:

- `>=` includes the starting date.
- `<=` includes the ending date.
- `&` means both conditions must be satisfied.
- Parentheses are required around individual conditions.

### Combining Filtering and Resampling

We can directly chain filtering and resampling without creating an additional DataFrame variable.

```python
df_sales[
    (df_sales["date"] >= "2026-11-01") &
    (df_sales["date"] <= "2026-12-31")
].resample("ME", on="date")["sales"].sum()
```

This performs two operations:

1. Selects records between November 1 and December 31.
2. Calculates the total sales separately for each month.

### Expected Output

| Month | Total Sales |
|---|---:|
| November | 11,800 |
| December | 17,000 |

**Important:** Resampling does not select specific months by itself. It groups the available data according to the specified frequency. Date filtering is used when we need a particular date range.

## 8. Calculating Average Monthly Sales

Sometimes we need average sales instead of total sales.

For this, we can use `.mean()` instead of `.sum()`.

```python
df_sales[
    (df_sales["date"] >= "2026-11-01") &
    (df_sales["date"] <= "2026-12-31")
].resample("ME", on="date")["sales"].mean()
```

### Expected Output

| Month | Average Sales |
|---|---:|
| November | 1,966.67 |
| December | 3,400.00 |

**Explanation:**

- November has 6 records: 11,800 / 6 = 1,966.67.
- December has 5 records: 17,000 / 5 = 3,400.

The `mean()` method calculates the average of the available records in each monthly group.

## 9. Weekly Sales Aggregation

We can also aggregate sales data weekly instead of monthly.

For this, we use the `"W"` frequency.

```python
df_sales[
    (df_sales["date"] >= "2026-11-01") &
    (df_sales["date"] <= "2026-12-31")
].resample("W", on="date")["sales"].sum()
```

### Understanding Weekly Resampling

By default, `"W"` groups data into weeks ending on Sunday.

For example:

- Monday to Sunday forms one weekly group.
- The next Monday to Sunday forms another weekly group.

**Important:** Weekly resampling uses calendar-week boundaries. If a date range begins or ends in the middle of a week, the first or last group may contain only the filtered records.

## 10. Different Aggregation Methods with Resampling

Resampling is not limited to calculating total sales.

We can use different aggregation methods depending on the analysis requirement.

| Method | Purpose |
|---|---|
| `.sum()` | Calculates total values |
| `.mean()` | Calculates average values |
| `.count()` | Counts non-missing records |
| `.max()` | Finds the maximum value |
| `.min()` | Finds the minimum value |

### Examples

**Monthly total sales:**

```python
df_sales.resample("ME", on="date")["sales"].sum()
```

**Monthly average sales:**

```python
df_sales.resample("ME", on="date")["sales"].mean()
```

**Monthly number of sales records:**

```python
df_sales.resample("ME", on="date")["sales"].count()
```

**Maximum monthly sales value:**

```python
df_sales.resample("ME", on="date")["sales"].max()
```

**Minimum monthly sales value:**

```python
df_sales.resample("ME", on="date")["sales"].min()
```

## 11. Important Concept: Resampling vs Filtering

These two operations have different purposes.

| Operation | Purpose |
|---|---|
| Filtering | Selects rows based on a condition |
| Resampling | Groups time-based records into intervals |
| Aggregation | Performs calculations on grouped records |

Example:

```python
df_sales[
    df_sales["date"] >= "2026-11-01"
]
```

This filters records starting from November 1.

Whereas:

```python
df_sales.resample("ME", on="date")["sales"].sum()
```

This calculates monthly total sales for the available date range.

We can combine both operations when required.

## 12. Method Chaining

Pandas allows us to combine multiple operations in a single expression.

Example:

```python
df_sales[
    (df_sales["date"] >= "2026-11-01") &
    (df_sales["date"] <= "2026-12-31")
].resample("ME", on="date")["sales"].sum()
```

Here, we perform:

1. Date filtering.
2. Monthly resampling.
3. Sales column selection.
4. Total calculation.

We do not need to create an additional DataFrame variable if we do not intend to reuse the filtered data.

## 13. Important Mistake to Avoid

Consider the following code:

```python
df_sales.resample("ME", on="date")["sales"].sum
```

This returns a reference to the method instead of executing it.

Correct:

```python
df_sales.resample("ME", on="date")["sales"].sum()
```

Remember to use parentheses when calling methods such as:

- `.sum()`
- `.mean()`
- `.count()`
- `.max()`
- `.min()`

4. `"D"` represents daily frequency.
5. `"W"` represents weekly frequency.

````python
#Resampling in Pandas
# aggregation of daily data on weekly, monthly level called RESAMPLING 

#SYNTAX: df.resample("frequency", on="date_column")["value_column"].sum()

#frequency → daily, weekly, monthly, etc.
#on → date column ka naam.
#value_column → jis numerical column ka calculation karna hai.
#sum() → total calculate karega.

data = {
    "date": [
        "2026-01-05", "2026-01-15", "2026-01-25",
        "2026-02-03", "2026-02-14",
        "2026-03-05", "2026-03-20",
        "2026-04-10", "2026-04-22",
        "2026-05-01", "2026-05-18",
        "2026-06-07", "2026-06-25",
        "2026-07-04", "2026-07-19",
        "2026-08-08", "2026-08-21",
        "2026-09-02", "2026-09-17",
        "2026-10-05", "2026-10-23",
        "2026-11-01", "2026-11-02", "2026-11-03",
        "2026-11-04", "2026-11-05", "2026-11-15",
        "2026-12-01", "2026-12-02", "2026-12-10",
        "2026-12-20", "2026-12-28"
    ],
    "sales": [
        1200, 1800, 2200,
        1500, 2500,
        1800, 3200,
        2100, 2900,
        1700, 2600,
        3000, 3500,
        2200, 2800,
        3100, 2700,
        1900, 3300,
        2500, 3600,
        1000, 2000, 1500, 3000, 2500, 1800,
        4000, 3500, 2200, 4500, 2800
    ]
}


df_sales = pd.DataFrame(data)

#converting date into date time format
df_sales["date"] = pd.to_datetime(df_sales["date"])

#calculate the total sales of january to december
df_sales[(df_sales["date"] >= "2026-01-01") & (df_sales["date"] <= "2026-12-31")
        ].resample("ME", on="date")["sales"].sum()

#calculate the sales of only novemeber to december 
df_sales[
    (df_sales["date"] >= "2026-11-01") &
    (df_sales["date"] <= "2026-12-31")
].resample("ME", on="date")["sales"].sum()

#calculate the average monthly sales of novemeber and december 
df_sales[
    (df_sales["date"] >= "2026-11-01") &
    (df_sales["date"] <= "2026-12-31")
].resample("ME", on="date")["sales"].mean()

#calculate the weekly sales of novemeber and december 
df_sales[
    (df_sales["date"] >= "2026-11-01") &
    (df_sales["date"] <= "2026-12-31")
].resample("W", on="date")["sales"].sum()
````

# Time Zones in Pandas

## 1. What are Time Zones?

A Time Zone is a geographical region that follows a particular standard time.

Different countries and regions have different local times because the Earth rotates and different locations experience daylight at different times.

For example:

- India follows Indian Standard Time (IST).
- London follows Greenwich Mean Time (GMT) during standard time.
- New York follows Eastern Standard Time (EST) during standard time.

### Why do we need Time Zones?

Imagine a person is working from India and another person is working from New York.

If both people have an online meeting at the same moment, their local clocks will show different times.

Time Zones help us understand and manage these differences.

They are especially useful in:

- International applications.
- Online meeting platforms.
- Flight booking systems.
- E-commerce applications.
- Server logs.
- Global transaction records.

## 2. What is UTC?

UTC stands for **Coordinated Universal Time**.

It is the primary reference time used to coordinate time around the world.

Different regions calculate their local time using an offset from UTC.

For example:

| Location | Standard Time Offset |
|---|---|
| UTC | UTC +00:00 |
| India (IST) | UTC +05:30 |
| London (GMT) | UTC +00:00 |
| New York (EST) | UTC -05:00 |

### Understanding UTC Offset

A UTC offset tells us how much a local time differs from UTC.

For example:

`UTC +05:30`

This means the local time is 5 hours and 30 minutes ahead of UTC.

Similarly:

`UTC -05:00`

This means the local time is 5 hours behind UTC.

**Important:** Some locations change their UTC offset during daylight saving time. Therefore, their offset can depend on the date.

## 3. What is a Timestamp in Pandas?

A Timestamp is a Pandas object used to represent a single date and time.

Pandas provides `pd.Timestamp()` to create a Timestamp object.

### Syntax

```python
pd.Timestamp("date and time")
```

### Example

```python
import pandas as pd

date = pd.Timestamp("2026-11-05 12:00:00")

print(date)
print(type(date))
```

Output:

```text
2026-11-05 12:00:00
<class 'pandas._libs.tslibs.timestamps.Timestamp'>
```

### Why use `pd.Timestamp()`?

It converts a single date-time value into a Pandas Timestamp object.

It allows us to perform Date-Time operations such as:

- Extracting year, month, and day.
- Comparing dates.
- Adding or subtracting durations.
- Working with Time Zones.

### `pd.Timestamp()` vs `pd.to_datetime()`

| `pd.Timestamp()` | `pd.to_datetime()` |
|---|---|
| Creates a single Timestamp object | Converts date-like values into Date-Time format |
| Useful for one date-time value | Useful for strings, lists, arrays, and Series |
| Returns a Timestamp for a single input | Return type depends on the input |

Examples:

```python
# Single date-time value
date = pd.Timestamp("2026-11-05 12:00:00")

# Multiple date values
dates = pd.to_datetime([
    "2026-11-05",
    "2026-11-06"
])
```

## 4. What is a Naive Datetime?

A Naive Datetime is a date-time value that does not contain any Time Zone information.

In simple words, Pandas knows the date and time, but it does not know which country's local time it represents.

### Example

```python
naive_date = pd.Timestamp("2026-11-05 14:30:00")

print(naive_date)
print(type(naive_date))

```

Output:

```text
2026-11-05 14:30:00
<class 'pandas._libs.tslibs.timestamps.Timestamp'>
None
```

### What does `None` mean?

The `.tz` attribute returns the Time Zone associated with a Timestamp.

If it returns `None`, no Time Zone is attached.

### Why does a Naive Datetime matter?

Suppose a database stores:

`2026-11-05 14:30:00`

Without Time Zone information, we cannot confidently determine whether this represents 2:30 PM in India, London, or New York.

This can cause problems when working with international applications or comparing timestamps from different locations.

## 5. What is a Timezone-Aware Datetime?

A Timezone-Aware Datetime is a date-time value that contains Time Zone information.

Pandas knows both the date-time value and the Time Zone to which it belongs.

### Syntax

```python
pd.Timestamp("date and time", tz="timezone")
```

### Example

```python
aware_date = pd.Timestamp(
    "2026-11-05 14:30:00",
    tz="Asia/Kolkata"
)

print(aware_date)
print(type(aware_date))
print(aware_date.tz)
```

Output:

```text
2026-11-05 14:30:00+05:30
<class 'pandas._libs.tslibs.timestamps.Timestamp'>
Asia/Kolkata
```

### Understanding the Output

`2026-11-05 14:30:00+05:30`

- `2026-11-05` represents the date.
- `14:30:00` represents the local time.
- `+05:30` represents the UTC offset.

`Asia/Kolkata` identifies the Time Zone.

### Why use Timezone-Aware Datetime?

It helps us:

- Identify the actual Time Zone of a timestamp.
- Compare timestamps from different regions.
- Convert local time into another region's time.
- Avoid confusion in international applications.

## 6. Naive vs Timezone-Aware Datetime

| Naive Datetime | Timezone-Aware Datetime |

| `.tz` returns `None` | `.tz` returns the associated Time Zone |

# Time Zone Localization and Conversion in Pandas

## 1. Introduction to Time Zone Operations

In real-world applications, date and time values may belong to different geographical regions.

For example:

- An employee works in India.
- The company server operates in UTC.
- A customer accesses the application from New York.

To handle these differences, Pandas provides two important methods:

1. `tz_localize()`
2. `tz_convert()`

Both methods work with Time Zone information, but their purposes are different.

---

## 2. What is `tz_localize()`?

### Definition

`tz_localize()` is used to attach Time Zone information to a Naive Datetime.

A Naive Datetime contains date and time information but does not have any associated Time Zone.

Using `tz_localize()`, we tell Pandas which Time Zone the existing date-time value belongs to.

### Why do we need `tz_localize()`?

Suppose an Indian company stores employee login times:

`2026-11-05 09:00:00`

The company knows that this time represents Indian local time.

However, Pandas does not know this automatically.

We use `tz_localize()` to attach the Indian Time Zone to the existing datetime.

### Syntax

For a single Timestamp:

```python
datetime.tz_localize("timezone")
```

For a Pandas Series:

```python
df["date"].dt.tz_localize("timezone")
```

### Example 1: Localizing a Single Timestamp

```python
import pandas as pd

naive_date = pd.Timestamp("2026-11-05 14:30:00")

aware_date = naive_date.tz_localize("Asia/Kolkata")

print(aware_date)
print(aware_date.tz)
```

Output:

```text
2026-11-05 14:30:00+05:30
Asia/Kolkata
```

### Understanding the Output

Before localization:

`2026-11-05 14:30:00`

After localization:

`2026-11-05 14:30:00+05:30`

Notice that the clock time remains unchanged.

Only the Time Zone information is attached.

### Example 2: Using `tz_localize()` in a DataFrame

```python
data = {
    "employee": ["Rahul", "Aman", "Priya"],
    "login_time": [
        "2026-11-05 09:00:00",
        "2026-11-05 09:30:00",
        "2026-11-05 10:00:00"
    ]
}

df_attendance = pd.DataFrame(data)

# Convert the column into Date-Time format
df_attendance["login_time"] = pd.to_datetime(
    df_attendance["login_time"]
)

# Attach the Indian Time Zone
df_attendance["login_time"] = (
    df_attendance["login_time"]
    .dt.tz_localize("Asia/Kolkata")
)

print(df_attendance)
print(df_attendance["login_time"].dtype)
```

The resulting column becomes timezone-aware.

Example dtype:

`datetime64[us, Asia/Kolkata]`

The resolution may differ depending on the Pandas version and datetime data.

### Important Points about `tz_localize()`

- It attaches Time Zone information to a Naive Datetime.
- It does not change the existing clock time.
- It is useful when we know which Time Zone the original datetime represents.
- It should not be applied again to an already timezone-aware datetime.
- For a Series, use the `.dt` accessor.

---

## 3. What is `tz_convert()`?

### Definition

`tz_convert()` is used to convert a Timezone-Aware Datetime from one Time Zone to another.

It changes the displayed local clock time while preserving the actual moment represented by the timestamp.

### Why do we need `tz_convert()`?

Suppose an employee logs in from India at:

`2026-11-05 14:30:00+05:30`

The company's central server uses UTC.

We need to convert the Indian local time into UTC.

Since India is 5 hours and 30 minutes ahead of UTC:

`14:30 - 05:30 = 09:00`

Therefore, the equivalent UTC time is 09:00 AM.

### Syntax

For a single Timestamp:

```python
datetime.tz_convert("target_timezone")
```

For a Pandas Series:

```python
df["date"].dt.tz_convert("target_timezone")
```

### Example 1: Converting India Time to UTC

```python
import pandas as pd

india_time = pd.Timestamp(
    "2026-11-05 14:30:00",
    tz="Asia/Kolkata"
)

utc_time = india_time.tz_convert("UTC")

print(india_time)
print(utc_time)
```

Output:

```text
2026-11-05 14:30:00+05:30
2026-11-05 09:00:00+00:00
```

Both timestamps represent the same actual moment.

Only their displayed local times and Time Zone information differ.

### Example 2: Converting India Time to New York Time

```python
new_york_time = india_time.tz_convert("America/New_York")

print(new_york_time)
```

Output for 5 November 2026:

```text
2026-11-05 04:00:00-05:00
```

Pandas automatically determines the appropriate New York offset for the given date.

### Example 3: Using `tz_convert()` in a DataFrame

```python
# Convert Indian local times into UTC
df_attendance["login_time"] = (
    df_attendance["login_time"]
    .dt.tz_convert("UTC")
)

print(df_attendance)
```

The original timezone-aware column is converted into UTC.

Example dtype:

`datetime64[us, UTC]`

### Important Points about `tz_convert()`

- It converts an existing timezone-aware datetime.
- It requires Time Zone information to be present.
- It preserves the actual moment represented by the timestamp.
- The displayed local clock time can change.
- For a Series, use the `.dt` accessor.
- For a single Timestamp, use `.tz_convert()` directly.

---

## 4. Difference Between `tz_localize()` and `tz_convert()`

This is one of the most important differences in Pandas Date-Time handling.

| Feature | `tz_localize()` | `tz_convert()` |
|---|---|---|
| Purpose | Attach a Time Zone | Convert to another Time Zone |
| Input | Naive Datetime | Timezone-Aware Datetime |
| Time Zone | Adds timezone information | Changes timezone information |
| Clock time | Remains unchanged | May change |
| Actual instant | Assigns an interpretation to the naive clock time | Preserves the existing instant |

### Visual Example

**Using `tz_localize()`:**

Before:

`2026-11-05 14:30:00`

After:

`2026-11-05 14:30:00+05:30`

The clock time remains 14:30.

**Using `tz_convert()`:**

Before:

`2026-11-05 14:30:00+05:30`

After:

`2026-11-05 09:00:00+00:00`

The clock time changes, but the actual moment remains the same.

### Remember This Rule

- `tz_localize()` → Attach Time Zone.
- `tz_convert()` → Convert Time Zone.

---

## 5. Common Time Zone Operations

| Operation | Syntax | Purpose |
|---|---|---|
| Create a timezone-aware Timestamp | `pd.Timestamp("2026-11-05", tz="UTC")` | Create a Timestamp with a Time Zone |
| Attach timezone | `date.tz_localize("Asia/Kolkata")` | Localize a Naive Datetime |
| Convert timezone | `date.tz_convert("UTC")` | Convert an aware datetime |
| Check timezone | `date.tz` | Get timezone information |
| Check dtype | `df["date"].dtype` | Inspect the datetime column dtype |

## 6. Common Mistakes

### Mistake 1: Using `tz_convert()` on a Naive Datetime

A Naive Datetime does not have timezone information.

Therefore, directly converting it using `tz_convert()` will raise an error.

First attach the correct timezone using `tz_localize()`.

### Mistake 2: Applying `tz_localize()` Twice

If a datetime is already timezone-aware, applying `tz_localize()` again without removing its timezone will raise an error.

Use `tz_convert()` if you want to change an existing timezone.

### Mistake 3: Forgetting to Assign the Result

Consider:

```python
df["date"].dt.tz_localize("Asia/Kolkata")
```

This returns a localized Series but does not update the original DataFrame column.

To update the column:

```python
df["date"] = df["date"].dt.tz_localize("Asia/Kolkata")
```

The same assignment principle applies to `tz_convert()`.

---

## 7. Practical Applications

Time Zone localization and conversion are useful in:

- **Attendance Systems:** Store employee login times with the correct local timezone.
- **E-commerce:** Handle orders placed from different countries.
- **Flight Booking:** Convert departure and arrival times into the appropriate local timezones.
- **Global Applications:** Display timestamps according to each user's location.
- **Server Logs:** Standardize timestamps using UTC.
- **Online Meetings:** Coordinate meetings across different countries.

## 8. Recommended Workflow

When working with a naive date-time column that represents a known local timezone, follow this sequence:

**Step 1:** Convert the column into Date-Time format.

```python
df["date"] = pd.to_datetime(df["date"])
```

**Step 2:** Attach the correct local timezone.

```python
df["date"] = df["date"].dt.tz_localize("Asia/Kolkata")
```

**Step 3:** Convert it into UTC if required.

```python
df["date"] = df["date"].dt.tz_convert("UTC")
```

**Important:** Never attach a timezone without knowing what timezone the original naive timestamps actually represent. Incorrect localization can produce incorrect time calculations.

---

## 9. Key Takeaways

1. `tz_localize()` attaches timezone information to a Naive Datetime.
2. `tz_convert()` converts an existing timezone-aware datetime into another timezone.
3. `tz_localize()` keeps the clock time unchanged.
4. `tz_convert()` preserves the actual moment but can change the displayed clock time.
5. Use `.dt` when applying these methods to a Pandas Series.
6. Use the methods directly when working with a single Timestamp.
7. UTC is commonly used as a standard timezone for storing and coordinating global timestamps.

**Final Rule to Remember:**

`Naive Datetime → tz_localize() → Timezone-Aware Datetime → tz_convert() → Another Time Zone`

# String Operations in Pandas

String operations are used to clean, modify, search, and manipulate text data in a Pandas DataFrame.

Pandas provides the `.str` accessor to perform string operations on an entire Series (column).

**General Syntax:**

```python
DataFrame["column_name"].str.function()
```

## 1. `str.lower()` – Convert to Lowercase

Converts all characters in a string into lowercase.

**Syntax:**
```python
df["column_name"].str.lower()
```

**Example:**
```python
df_raw["name"].str.lower()
```

**Output:**
```text
priyam shankar
rahul kumar
ankit sharma
```

**Use Case:** Standardizing text values that have inconsistent capitalization.

## 2. `str.upper()` – Convert to Uppercase

Converts all characters in a string into uppercase.

**Syntax:**
```python
df["column_name"].str.upper()
```

**Example:**
```python
df_raw["gender"].str.upper()
```

**Output:**
```text
MALE
FEMALE
MALE
```

**Use Case:** Standardizing categorical values such as gender, country codes, and state codes.

## 3. `str.capitalize()` – Capitalize First Character

Converts the first character of a string into uppercase and all remaining characters into lowercase.

**Syntax:**
```python
df["column_name"].str.capitalize()
```

**Example:**
```python
df_raw["name"].str.capitalize()
```

**Output:**
```text
Priyam shankar
Rahul kumar
Ankit sharma
```

**Important:** `capitalize()` capitalizes only the first character of the entire string, not the first character of every word.

To capitalize the first character of every word, use:

```python
df_raw["name"].str.title()
```

**Output:**
```text
Priyam Shankar
Rahul Kumar
Ankit Sharma
```

## 4. `str.strip()` – Remove Extra Spaces

Removes unnecessary spaces from the beginning (leading spaces) and end (trailing spaces) of a string.

**Syntax:**
```python
df["column_name"].str.strip()
```

**Example:**
```python
df_raw["name"].str.strip()
```

**Example:**

Before:
```text
"   Priyam Shankar   "
```

After:
```text
"Priyam Shankar"
```

**Related Functions:**

- `str.lstrip()` – Removes leading spaces.
- `str.rstrip()` – Removes trailing spaces.
- `str.strip()` – Removes both leading and trailing spaces.

**Use Case:** Cleaning text data imported from CSV or Excel files.

## 5. `str.split()` – Split Strings into Parts

Splits a string into multiple parts based on a specified separator.

**Syntax:**
```python
df["column_name"].str.split("separator")
```

**Example:**
```python
df_raw["name"].str.split(" ")
```

Here, `" "` represents a space separator.

**Output:**
```text
["Priyam", "Shankar"]
["Rahul", "Kumar"]
["Ankit", "Sharma"]
```

**Using `expand=True`:**

The `expand=True` parameter converts the split values into separate DataFrame columns.

```python
df_raw["name"].str.split(" ", expand=True)
```

**Output:**

| First Part | Second Part |
|---|---|
| Priyam | Shankar |
| Rahul | Kumar |
| Ankit | Sharma |

**Use Case:** Separating full names into first name and last name.

## 6. `str.contains()` – Check Whether a String Exists

Checks whether a particular substring or pattern exists inside a string.

Returns Boolean values (`True` or `False`).

**Syntax:**
```python
df["column_name"].str.contains("substring")
```

**Example:**
```python
df_raw["country"].str.contains("US")
```

**Output:**
```text
False
True
True
```

**Case Sensitivity:**

By default, `str.contains()` is case-sensitive.

Example:
```python
df_raw["country"].str.contains("india")
```

This may return `False` for `"India"` because uppercase `I` and lowercase `i` are different.

**Ignoring Case:**

```python
df_raw["country"].str.contains("india", case=False)
```

The `case=False` parameter ignores uppercase and lowercase differences.

**Use Case:** Searching and filtering records based on specific text values.

Example:
```python
df_raw[df_raw["country"].str.contains("india", case=False)]
```

This returns rows where the country column contains "india", regardless of capitalization.

---

## Combining String Operations

Multiple string operations can be chained together to perform multiple cleaning tasks.

**Example:**

```python
df_raw["country"].str.strip().str.lower()
```

# Data Transformation in Pandas – Feature Engineering

## 1. Introduction to Feature Engineering

**Feature Engineering** is the process of transforming existing data or creating new columns (features) from existing data to improve the performance of a Machine Learning model.

In simple words, we modify our existing data or create new features to make it more useful for analysis and Machine Learning.

**Example:**

Suppose we have an employee dataset containing:

- Employee ID
- Name
- Age
- Gender
- Country
- Income

We can create additional features such as:

- Tax – Calculate tax based on employee income.
- New Income – Calculate income after a salary increase.
- Ranking – Rank employees according to their income.

These newly created columns are called **features**.

---

## 3. `apply()` – Apply Custom Logic to Data

The `apply()` function is used when we want to apply our own logic or a custom function to existing data.

It is useful when we want to transform values based on certain conditions.

**Example: Creating a Tax Column**

Suppose we have an employee dataset, and we want to calculate tax based on income:

- Employees earning 60,000 or more will have a 20% tax.
- Employees earning less than 60,000 will have a 10% tax.

First, create a copy of the original DataFrame:

```python
df2_raw = df_raw.copy()
```

Now, create a new column called `tax` using `apply()`:

```python
df2_raw["tax"] = df2_raw["income"].apply(
    lambda x: "20%" if x >= 60000 else "10%"
)
```

**Explanation:**

- `apply()` applies the given logic to each value in the column.
- `lambda x` represents each individual income value.
- `if x >= 60000` checks whether income is greater than or equal to 60,000.
- If the condition is True, it returns `"20%"`.
- Otherwise, it returns `"10%"`.

**Important:** We use a lambda function here to write a small function in a single line without defining it separately using `def`.

**Use Case:** Applying custom conditions, calculations, and transformations to existing columns.

---

## 4. `map()` – Replace Existing Values Using a Mapping

The `map()` function is used when we want to assign new values to existing values in a Series.

It works with a dictionary that defines the relationship between old values and new values.

**Example: Converting Gender Values**

Suppose our gender column contains:

- Male
- Female
- Unknown

We want to convert them into:

- Male → M
- Female → F
- Unknown → U

First, create a dictionary:

```python
gender_map = {
    "Male": "M",
    "Female": "F",
    "Unknown": "U"
}
```

Now, apply the mapping:

```python
df2_raw["gender"] = df2_raw["gender"].map(gender_map)
```

**Explanation:**

- `map()` checks each value in the gender column.
- It searches for the corresponding value in the dictionary.
- If a matching key exists, it replaces the value with its mapped value.

**Important:** If a value does not have a matching key in the dictionary, `map()` returns `NaN` for that value.

**Use Case:** Converting categorical values into short labels or standardized values.

---

## 5. `assign()` – Create New Columns

The `assign()` function is used to create new columns in a DataFrame by assigning calculated or existing values.

**Example: Increasing Employee Income by 10%**

Suppose every employee's income increases by 10%, and we want to store the increased income in a new column called `new_income`.

```python
df2_raw = df2_raw.assign(
    new_income=df2_raw["income"] * 1.1
)
```

**Explanation:**

- `assign()` creates a new column.
- `df2_raw["income"]` accesses the existing income column.
- Multiplying by `1.1` increases the income by 10%.
- The calculated values are stored in `new_income`.

**Example:**

If the original income is 50,000:

New Income = 50,000 × 1.1 = 55,000

**Important:** `assign()` returns a new DataFrame. Therefore, we assign the result back to `df2_raw` to retain the changes.

**Use Case:** Creating new calculated columns such as increased salary, total price, profit, or age categories.

## 6. `replace()` – Replace Specific Values

The `replace()` function is used to replace specific old values with new values.

**Syntax:**

```python
DataFrame["column_name"].replace(old_value, new_value)
```

Suppose our country column contains `"USA"`, and we want to replace it with `"United States"`.

```python
df2_raw["country"] = df2_raw["country"].replace(
    "USA", "United States"
)
```

**Explanation:**

- `"USA"` is the old value.
- `"United States"` is the new value.
- Every matching `"USA"` value in the column is replaced.

**Use Case:** Correcting inconsistent values, replacing abbreviations, and cleaning categorical data.

**Explanation:**

**Important:** `rename()` returns a new DataFrame by default. To retain the changes, assign the result back to a variable or use `inplace=True`.

Sorting helps us arrange data in a particular order based on column values or index labels.

- `sort_values()`
- `sort_index()`

This sorts only the Income Series in ascending order by default.

**Example 2: Sorting the Entire DataFrame**


This sorts the complete DataFrame according to the Income column.

By default, sorting happens in ascending order (lowest to highest).

To sort in descending order:

```python
df2_raw.sort_values("Income", ascending=False)
```

- `drop=True` prevents the old index from being added as a new column.

**Use Case:** Organizing DataFrames after sorting, filtering, or other operations.

## 9. `rank()` – Assign Ranks to Data

The `rank()` function assigns numerical ranks to values in a Series.

It is useful when we want to rank employees, students, products, or other records based on a particular parameter.

**Example: Creating a Ranking Column Based on Income**

```python
sorted_df["Ranking"] = sorted_df["Income"].rank()
```

**Explanation:**

- `rank()` assigns a numerical rank to each income value.
- By default, the lowest value receives rank 1.
- The highest value receives the largest rank.
- If two values are equal, their ranks are averaged by default.
- Missing values (`NaN`) receive `NaN` ranks by default.

### A. Ranking in Descending Order

By default, ranking starts from the lowest value.

If we want the highest income to receive rank 1, we use:

```python
sorted_df["Ranking"] = sorted_df["Income"].rank(
    ascending=False
)
```

### B. Handling Tied Ranks Using `method`

When two or more values are equal, Pandas provides different methods to assign ranks.

**1. Average (Default)**

```python
sorted_df["Income"].rank(method="average")
```

Assigns the average of the ranks occupied by tied values.

**2. Dense Ranking**

```python
sorted_df["Income"].rank(method="dense")
```

Assigns the same rank to tied values without skipping the next rank.

**3. Minimum Ranking**

```python
sorted_df["Income"].rank(method="min")
```

Assigns the minimum rank among the tied positions.

**4. Maximum Ranking**

```python
sorted_df["Income"].rank(method="max")
```

Assigns the maximum rank among the tied positions.

**Example:**

Suppose we have the following income values:

| Income | Average | Dense | Min | Max |
|---:|---:|---:|---:|---:|
| 30000 | 1.0 | 1.0 | 1.0 | 1.0 |
| 40000 | 2.5 | 2.0 | 2.0 | 3.0 |
| 40000 | 2.5 | 2.0 | 2.0 | 3.0 |
| 50000 | 4.0 | 3.0 | 4.0 | 4.0 |

**Important:** `rank(method="dense")` does not skip ranks after tied values.

**Use Case:** Ranking students according to marks, employees according to income, or products according to sales.

2. Handling missing values.
3. Sorting the data.
4. Resetting the index.

```python
df2_raw = df_raw.copy()
```

```python
df2_raw = df2_raw.drop_duplicates()
```

```python
df2_raw = df2_raw.fillna(0)
```

```python
df2_raw = df2_raw.sort_values("income")
```

```python
df2_raw = df2_raw.reset_index(drop=True)
```

# Grouping and Aggregation of Data in Pandas

## 1. Introduction

Grouping and Aggregation are used to organize data into different groups and calculate meaningful summary statistics for each group.

**Why do we need Grouping and Aggregation?**

Suppose we have an employee dataset containing:

- Country
- Gender
- Age
- Income

Now, we want to answer questions like:

- What is the average income of employees in each country?
- What is the minimum and maximum income in each country?
- What is the average income of males and females?
- What is the average age of each gender?

Instead of filtering the data manually, we can use `groupby()` and aggregation functions.

## 2. `groupby()` – Group Data Based on a Column

The `groupby()` method divides the DataFrame into different groups based on the values of a particular column.

**General Syntax:**

```python
df.groupby("grouping_column")["target_column"].aggregation_function()
```

Where:

- `grouping_column` – The column used to create groups.
- `target_column` – The column on which we want to perform calculations.
- `aggregation_function()` – The calculation we want to perform.

### Example 1: Average Income Based on Country

Suppose we want to calculate the average income of employees in each country.

```python
df_raw.groupby("country")["income"].mean()
```

**Explanation:**

1. `groupby("country")` groups employees according to their country.
2. `["income"]` selects the income column.
3. `.mean()` calculates the average income for each country.

### Example 2: Minimum Income Based on Country

```python
df_raw.groupby("country")["income"].min()
```

This calculates the minimum income in each country.

### Example 3: Maximum Income Based on Country

```python
df_raw.groupby("country")["income"].max()
```

This calculates the maximum income in each country.

### Example 4: Average Income Based on Gender

```python
df_raw.groupby("gender")["income"].mean()
```

This calculates the average income of each gender.

### Example 5: Average Age Based on Gender

```python
df_raw.groupby("gender")["age"].mean()
```

This calculates the average age of each gender.

## 3. Aggregation Functions

Aggregation functions are used to calculate summary values from grouped data.

Some commonly used aggregation functions are:

| Function | Purpose |
|---|---|
| `mean()` | Calculates average |
| `sum()` | Calculates total |
| `min()` | Finds minimum value |
| `max()` | Finds maximum value |
| `count()` | Counts non-missing values |
| `median()` | Calculates median |
| `std()` | Calculates standard deviation |

**Example:**

```python
df_raw.groupby("country")["income"].mean()
```

Here, `mean()` is the aggregation function.

- Minimum Income

# Melt and Pivot in Pandas – Data Reshaping

## 1. Introduction

In Pandas, sometimes we need to change the structure or format of our DataFrame without changing the actual data.

This process is called **Data Reshaping**.

Pandas provides two important methods for reshaping data:

1. `melt()` – Converts Wide Format into Long Format.
2. `pivot()` – Converts Long Format into Wide Format.

## 2. Understanding Wide and Long Format

### A. Wide Format

In Wide Format, data is spread across multiple columns.

**Example:**

| Country | Year | Sales | Profit |
|---|---:|---:|---:|
| USA | 2020 | 100 | 20 |
| USA | 2021 | 120 | 25 |
| India | 2020 | 90 | 18 |
| India | 2021 | 110 | 22 |

Here, Sales and Profit are stored in separate columns.

### B. Long Format

In Long Format, multiple columns are converted into two columns:

- One column stores the original column names.
- Another column stores their corresponding values.

**Example:**

| Country | Year | Metrics | Value |
|---|---:|---|---:|
| USA | 2020 | Sales | 100 |
| USA | 2020 | Profit | 20 |
| USA | 2021 | Sales | 120 |
| USA | 2021 | Profit | 25 |
| India | 2020 | Sales | 90 |
| India | 2020 | Profit | 18 |
| India | 2021 | Sales | 110 |
| India | 2021 | Profit | 22 |

Here, Sales and Profit are combined into a single column called `Metrics`, and their corresponding values are stored in the `Value` column.

---

## 3. `melt()` – Convert Wide Format into Long Format

The `melt()` function is used to transform a DataFrame from Wide Format into Long Format.

It combines multiple columns into two columns while keeping selected identifier columns unchanged.

### Syntax

```python
pd.melt(
    df,
    id_vars,
    value_vars,
    var_name,
    value_name
)
```

### Parameters

| Parameter | Purpose |
|---|---|
| `id_vars` | Columns that we want to keep unchanged |
| `value_vars` | Columns that we want to melt into a single column |
| `var_name` | Name of the new column containing the original column names |
| `value_name` | Name of the new column containing the corresponding values |

### Example: Applying `melt()`

First, create a DataFrame:

```python
data = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22]
})
```

Now, convert Wide Format into Long Format:

```python
melted_data = data.melt(
    id_vars=["country", "year"],
    value_vars=["sales", "profit"],
    var_name="metrics",
    value_name="value"
)
```

### Explanation

- `id_vars=["country", "year"]` – Keeps country and year unchanged.
- `value_vars=["sales", "profit"]` – Combines sales and profit columns.
- `var_name="metrics"` – Stores the original column names (sales and profit).
- `value_name="value"` – Stores their corresponding values.

**Result:** Two separate columns (sales and profit) are transformed into a single column containing their names and another column containing their values.

---

## 4. `pivot()` – Convert Long Format into Wide Format

The `pivot()` function performs the opposite operation of `melt()`.

It converts Long Format data into Wide Format by spreading unique values from one column into separate columns.

### Syntax

```python
df.pivot(
    index,
    columns,
    values
)
```

### Parameters

| Parameter | Purpose |
|---|---|
| `index` | Column(s) used to identify the rows in the new DataFrame |
| `columns` | Column whose unique values become new column names |
| `values` | Column containing the actual values to fill into the new DataFrame |

### Example: Applying `pivot()`

Suppose we have already converted our original DataFrame into Long Format using `melt()`.

Now, we want to convert it back into Wide Format.

```python
melted_data.pivot(
    index=["country", "year"],
    columns="metrics",
    values="value"
)
```

### Explanation

- `index=["country", "year"]` – These columns identify each row.
- `columns="metrics"` – Unique values (sales and profit) become separate columns.
- `values="value"` – Fills the new columns with their corresponding values.

**Result:** We get the original Wide Format structure with separate Sales and Profit columns.

**Important:** `pivot()` requires each combination of index and column values to be unique. If duplicate combinations exist, `pivot()` raises an error.

---

## 5. Difference Between `melt()` and `pivot()`

| Feature | `melt()` | `pivot()` |
|---|---|---|
| Transformation | Wide to Long | Long to Wide |
| Purpose | Combines multiple columns | Spreads values into multiple columns |
| Structure | More rows, fewer columns | Fewer rows, more columns |
| Main parameters | `id_vars`, `value_vars` | `index`, `columns`, `values` |
| Use | Data normalization and reshaping | Restoring a wider structure |

## 6. Quick Revision

**MELT = Wide → Long**

```python
melted_data = data.melt(
    id_vars=["country", "year"],
    value_vars=["sales", "profit"],
    var_name="metrics",
    value_name="value"
)
```

**PIVOT = Long → Wide**

```python
melted_data.pivot(
    index=["country", "year"],
    columns="metrics",
    values="value"
)
```

## 7. Key Takeaways

- Data Reshaping means changing the structure of a DataFrame.
- Wide Format stores different variables in separate columns.
- Long Format stores variable names and their corresponding values in separate columns.
- `melt()` converts Wide Format into Long Format.
- `pivot()` converts Long Format into Wide Format.
- `id_vars` specifies the columns that remain unchanged during melting.
- `value_vars` specifies the columns that will be combined.
- `index`, `columns`, and `values` control how data is arranged during pivoting.

**Remember:**

`melt()` = Wide to Long

`pivot()` = Long to Wide

Both methods are useful when preparing datasets for data analysis, visualization, and Machine Learning.

````python
#  MELT &  PIVOT

#MELT : Wide format data --> convert --> Long format data;    Eg: we have 10 columns & 3 rows data ---> wide range data --> convert --> long range data

#use melt method -->  SYNTAX:  melt(1stparameter, 2ndparameter, 3rdparameter, 4thparameter)
# SYNTAX:  melt(id_vars, value_vars, var_name, value_name)

#PARAMETRS
#1. id_vars : columns we want to keep    Eg: from our df --> we want--> country & year --> keep ---> we pass these 2 into a list 
#2. value_vars : these are column which we want to melt into one single column   Eg: we want --> sales & profit --> melt into 1 --> we pass these 2 into a list
#3. var_name : what is the name we give to the melted columns   EG: we melts 2 col: sales & profit --> let name is "matrics" --> ="matrics"
#4. value_name : what is the name of the new column where value is store 

data = pd.DataFrame({
    "country" : ["USA", "USA", "India", "India"],
    "year" : [2020, 2021, 2020, 2021],
    "sales" : [100, 120, 90, 110],
    "profit" : [20, 25, 18, 22]
})


#By this we melt our data, we transform our data 
melted_data = data.melt(
    id_vars=["country", "year"],
    value_vars=["sales", "profit"],
    var_name="matrics",
    value_name="value"
)


# Now if we've to do its opposite then we use PIVOT
#PIVOT : long to wide fromat

#SYNTAX:  pivot(index, columns, values)

#1. index : the new rows which formed for our new data format is index 
melted_data.pivot(
    index=["country", "year"],
    columns="matrics",
    values="value"
)
````

# Merging and Joining Data in Pandas

## 1. Introduction

In real-world projects, data is often stored in multiple tables or DataFrames instead of a single table.

Sometimes, we need to combine information from different DataFrames to get a complete dataset.

This process is called **Merging or Joining Data**.

### Why Do We Need Merging and Joining?

Suppose we have two separate DataFrames:

**Customers DataFrame:**
- Customer ID
- Customer Name

**Orders DataFrame:**
- Order ID
- Customer ID
- Order Amount

Now, suppose we want to know:

- Which customer placed which order?
- What is the name of the customer who placed an order?
- What is the order amount for each customer?

We need to combine both DataFrames using their common column, `customer_id`.

For this purpose, Pandas provides the `merge()` function.

---

## 2. `pd.merge()` – Combine Two DataFrames

The `merge()` function combines two DataFrames based on one or more common columns.

### General Syntax

```python
pd.merge(
    df1,
    df2,
    on="common_column",
    how="join_type"
)
```

### Parameters

| Parameter | Purpose |
|---|---|
| `df1` | First DataFrame |
| `df2` | Second DataFrame |
| `on` | Common column used to combine both DataFrames |
| `how` | Specifies the type of join |

**Important:** By default, `how="inner"` is used.

---

## 3. Creating Example DataFrames

Let's create two DataFrames to understand all four types of joins.

### Customers DataFrame

```python
df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})
```

| customer_id | name |
|---:|---|
| 1 | Adam |
| 2 | Bob |
| 3 | Charlie |
| 4 | Dave |

### Orders DataFrame

```python
df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [205, 120, 300, 180]
})
```

| order_id | customer_id | amount |
|---:|---:|---:|
| 101 | 2 | 205 |
| 102 | 1 | 120 |
| 103 | 4 | 300 |
| 104 | 5 | 180 |

Notice:

- Customer IDs 1, 2, and 4 exist in both DataFrames.
- Customer ID 3 exists only in `df_customers`.
- Customer ID 5 exists only in `df_orders`.

Now let's understand the four types of joins.

---

## 4. Types of Joins in Pandas

Pandas provides four commonly used join types:

1. Inner Join
2. Left Join
3. Right Join
4. Outer Join

### A. Inner Join (Default)

An Inner Join returns only the rows that have matching values in both DataFrames.

**Syntax:**

```python
pd.merge(
    df_customers,
    df_orders,
    on="customer_id"
)
```

**Or explicitly:**

```python
pd.merge(
    df_customers,
    df_orders,
    on="customer_id",
    how="inner"
)
```

**Output:**

| customer_id | name | order_id | amount |
|---:|---|---:|---:|
| 1 | Adam | 102 | 120 |
| 2 | Bob | 101 | 205 |
| 4 | Dave | 103 | 300 |

**Explanation:**

- Only matching customer IDs are included.
- Customer ID 3 is excluded because it has no matching order.
- Customer ID 5 is excluded because it has no matching customer.

**Remember:** Inner Join = Only Matching Rows.

### B. Left Join

A Left Join returns all rows from the left DataFrame and matching rows from the right DataFrame.

If a matching value does not exist in the right DataFrame, the corresponding columns contain `NaN`.

**Syntax:**

```python
pd.merge(
    df_customers,
    df_orders,
    on="customer_id",
    how="left"
)
```

**Output:**

| customer_id | name | order_id | amount |
|---:|---|---:|---:|
| 1 | Adam | 102 | 120 |
| 2 | Bob | 101 | 205 |
| 3 | Charlie | NaN | NaN |
| 4 | Dave | 103 | 300 |

**Explanation:**

- All customer records are included.
- Matching order details are added.
- Customer ID 3 has no matching order, so order details are `NaN`.
- Customer ID 5 is excluded because it exists only in the right DataFrame.

**Remember:** Left Join = All Left Table Rows + Matching Right Table Rows.

### C. Right Join

A Right Join returns all rows from the right DataFrame and matching rows from the left DataFrame.

If a matching value does not exist in the left DataFrame, the corresponding columns contain `NaN`.

**Syntax:**

```python
pd.merge(
    df_customers,
    df_orders,
    on="customer_id",
    how="right"
)
```

**Output:**

| customer_id | name | order_id | amount |
|---:|---|---:|---:|
| 2 | Bob | 101 | 205 |
| 1 | Adam | 102 | 120 |
| 4 | Dave | 103 | 300 |
| 5 | NaN | 104 | 180 |

**Explanation:**

- All order records are included.
- Matching customer names are added.
- Customer ID 5 has no matching customer, so the name is `NaN`.
- Customer ID 3 is excluded because it exists only in the left DataFrame.

**Remember:** Right Join = All Right Table Rows + Matching Left Table Rows.

### D. Outer Join (Full Outer Join)

An Outer Join returns all rows from both DataFrames.

If a matching value exists, the rows are combined. Otherwise, missing values are filled with `NaN`.

**Syntax:**

```python
pd.merge(
    df_customers,
    df_orders,
    on="customer_id",
    how="outer"
)
```

**Output:**

| customer_id | name | order_id | amount |
|---:|---|---:|---:|
| 1 | Adam | 102 | 120 |
| 2 | Bob | 101 | 205 |
| 3 | Charlie | NaN | NaN |
| 4 | Dave | 103 | 300 |
| 5 | NaN | 104 | 180 |

**Explanation:**

- All customer records are included.
- All order records are included.
- Matching records are combined.
- Non-matching records are also retained with `NaN` where information is missing.

**Remember:** Outer Join = All Rows from Both DataFrames.

---

## 5. Difference Between All Four Joins

| Join Type | What Does It Return? |
|---|---|
| Inner Join | Only matching rows |
| Left Join | All left rows + matching right rows |
| Right Join | All right rows + matching left rows |
| Outer Join | All rows from both DataFrames |

### Easy Way to Remember

- **Inner:** Only common records.
- **Left:** Everything from the first DataFrame.
- **Right:** Everything from the second DataFrame.
- **Outer:** Everything from both DataFrames.

# Data Concatenation in Pandas

## 1. Introduction to Data Concatenation

**Data Concatenation** is the process of combining multiple DataFrames by stacking them on top of each other (row-wise) or placing them side by side (column-wise).

In simple words, concatenation helps us combine multiple DataFrames into a single DataFrame.

### Why Do We Need Data Concatenation?

In real-world applications, data is not always available in a single DataFrame.

Sometimes, we fetch data in multiple parts because of:

- API rate limits.
- Database queries.
- Large datasets.
- Data collected from different sources.

Each part of the data may be stored in a separate DataFrame.

Eventually, we need to combine all these DataFrames into one complete dataset.

For this purpose, Pandas provides the `concat()` function.

**Real-Life Example:**

Suppose an API allows us to fetch only 1,000 records per request.

If we need 3,000 records, we may fetch them in three separate requests and store them in three DataFrames.

We can then use `concat()` to combine all three DataFrames into one.

---

## 2. `pd.concat()` – Combine Multiple DataFrames

The `concat()` function is used to combine two or more DataFrames along rows or columns.

### General Syntax

```python
pd.concat([df1, df2], axis=0)
```

**Important Parameters:**

| Parameter | Purpose |
|---|---|
| `[df1, df2]` | List of DataFrames to combine |
| `axis=0` | Combines DataFrames row-wise (default) |
| `axis=1` | Combines DataFrames column-wise |
| `ignore_index=True` | Creates a fresh sequential index |

There are two main types of Data Concatenation:

1. Row Concatenation
2. Column Concatenation

---

## 3. Row Concatenation (Vertical Concatenation)

Row Concatenation means stacking one DataFrame on top of another.

In this process:

- The first DataFrame appears at the top.
- The second DataFrame appears below it.
- Rows are combined vertically.

### Example

First, create two DataFrames:

**DataFrame 1:**

```python
df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Priyam", "Priti", "Supriya"]
})
```

**DataFrame 2:**

```python
df2 = pd.DataFrame({
    "id": [4, 5, 6],
    "name": ["Subodh", "Amit", "Rupak"]
})
```

### Applying Row Concatenation

```python
pd.concat([df1, df2])
```

**Output:**

| Index | id | name |
|---:|---:|---|
| 0 | 1 | Priyam |
| 1 | 2 | Priti |
| 2 | 3 | Supriya |
| 0 | 4 | Subodh |
| 1 | 5 | Amit |
| 2 | 6 | Rupak |

**Explanation:**

- Both DataFrames are stacked vertically.
- All rows from both DataFrames are included.
- By default, the original index labels are preserved.

### Resetting the Index Using `ignore_index=True`

Notice that the output contains repeated index values (0, 1, 2).

If we want a fresh sequential index, we use `ignore_index=True`.

```python
pd.concat([df1, df2], ignore_index=True)
```

**Output:**

| Index | id | name |
|---:|---:|---|
| 0 | 1 | Priyam |
| 1 | 2 | Priti |
| 2 | 3 | Supriya |
| 3 | 4 | Subodh |
| 4 | 5 | Amit |
| 5 | 6 | Rupak |

**Use Case:** Combining data fetched from multiple API requests or different batches of a dataset.

---

## 4. Column Concatenation (Horizontal Concatenation)

Column Concatenation means placing one DataFrame beside another.

In this process:

- The first DataFrame appears on the left.
- The second DataFrame appears on the right.
- Columns are combined horizontally.

### Example

```python
pd.concat([df1, df2], axis=1)
```

**Output:**

| Index | id | name | id | name |
|---:|---:|---|---:|---|
| 0 | 1 | Priyam | 4 | Subodh |
| 1 | 2 | Priti | 5 | Amit |
| 2 | 3 | Supriya | 6 | Rupak |

**Explanation:**

- `axis=1` tells Pandas to concatenate along columns.
- Both DataFrames are placed side by side.
- By default, Pandas aligns rows using their index labels.

**Use Case:** Combining different sets of columns belonging to the same indexed records.

**Important:** Column concatenation does not automatically match records using an ID column. It aligns DataFrames using their index labels.

---

## 5. Difference Between Row and Column Concatenation

| Feature | Row Concatenation | Column Concatenation |
|---|---|---|
| Direction | Vertical | Horizontal |
| Axis | `axis=0` | `axis=1` |
| Operation | Stacks rows | Places columns side by side |
| Default | Yes | No |
| Common use | Combining data batches | Combining different columns |

---

## 6. Quick Revision

**1. Row Concatenation:**

```python
pd.concat([df1, df2])
```

- Combines DataFrames vertically.
- Preserves original index labels by default.

**2. Row Concatenation with Fresh Index:**

```python
pd.concat([df1, df2], ignore_index=True)
```

- Combines DataFrames vertically.
- Creates a fresh sequential index.

**3. Column Concatenation:**

```python
pd.concat([df1, df2], axis=1)
```

- Combines DataFrames horizontally.
- Aligns rows using index labels.

---

## 7. Key Takeaways

- Concatenation combines multiple DataFrames into one.
- `pd.concat()` is used for Data Concatenation.
- `axis=0` performs row-wise concatenation.
- `axis=1` performs column-wise concatenation.
- By default, original index labels are preserved.
- `ignore_index=True` creates a fresh sequential index.
- Column concatenation aligns data using index labels.

**Remember:**

- `axis=0` = Top to Bottom (Rows).
- `axis=1` = Left to Right (Columns).
- `ignore_index=True` = Fresh Index.

**Overall:** Data Concatenation is especially useful when working with large datasets, APIs, databases, and data collected in multiple batches.
